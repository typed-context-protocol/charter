## **Why Now: The Rise of Context**

**Context coordinates understanding.** This means the ability to perceive where conflicts and knowledge gaps exist, as well as navigate the optimal way to fill in knowledge gaps, relative to users' perspectives.

**TECHNICAL DEFINITION**: Transformation of insufficient understanding state to sufficient understanding state. That's a local experience that simulataneously exists with global perspectives.

**DISTRIBUTED COGNITIVE NETWORKS**: Benefits of local interactions can be mined for predictions and actions from global system. Address-based URIs are required for every CGP record entry.

But because of the volume and speed with which AI can produce code, data, and content, it is difficult for humans to know where and how to ensure Context is properly managed.

The Context Graph Protocol (`CGP` for short) introduces four _```facets```_, the first is `Data` that is being observed, and the other three are `Meaning`, `Structure`, and `World`.

Secondly, CGP introduces multiple layers that handle `intent` and `interpretation`: the `user`'s, the `system`'s, and the `business`.

Across these two categories, conflicts between interpretation and intent can exist, whether they are explicitly known or unknown, that can effect decision-making. This notebook aims to make the costs of those decisions explicit.

A key challenge is that Ontologists primarily focus on the ```Meaning``` Facet, ```Structure``` is mostly handled by the API Engineering community, and ```World``` Facet concerns are primarily owned by the business community and the users of their systems (employees and users).

A Context Graph brings perspectives together, to identify and minimize the uncertainty that people, systems, and AI don't have the info they need to make optimal decisions.

## **The Context Gauge: Measuring the Conditions Required for Interpretation**

Design tools provide rulers, grids, snapping, and alignment guides because unaided judgment is not sufficient to draw a perfectly straight line or verify that two objects are aligned.

Those tools do not decide what the designer should create. They provide stable references against which the designer's intent and the resulting geometry can be inspected.

The Context Graph Protocol serves the same purpose for measuring whether an immediate sender’s intent and a reader’s interpretation remain coherent when Data crosses a boundary.

> **The Context Graph Protocol does not decide what Data should mean, how it should be structured, or what it should refer to. It provides a gauge for declaring, carrying, and comparing those conditions at every handoff.**

The gauge measures three Context Facets:

| Context Facet | Question answered | Examples |
| --- | --- | --- |
| `Meaning` | What does the Data signify? | Celsius or Fahrenheit; gross or net revenue; customer or account holder |
| `Structure` | How must the Data be encoded, organized, generated, filtered, or validated? | schema version; date format; API filter; generator rule; parser; constraint validator |
| `World` | What must the interpreter know about the real-world situation to use the datum operationally? | a particular vessel; the Baltic Sea; fiscal year 2026; eligible customers; the currently authorized policy |

These are three distinct dimensions of interpretation.

#### Meaning

The `Meaning` Facet identifies the semantic interpretation of the Data.

For example:

```text
temperature = 28
```

does not establish whether `28` means Celsius or Fahrenheit.

The Data remains unchanged while its meaning changes.

#### Structure

The `Structure` Facet identifies the rules required to correctly encode, decode, organize, generate, filter, or validate the Data.

For example:

```text
03/04/2026
```

may be structurally interpreted as:

```text
DD/MM/YYYY
```

or:

```text
MM/DD/YYYY
```

Structure can also include:

* the schema against which a record is validated;
* the filter applied by an API;
* the generator that produced a value;
* the parser used to decode it;
* the treatment of missing values;
* aggregation and rounding rules; and
* constraints determining which states are valid.

Two systems can agree on the meaning of a field while applying different structural rules and therefore construct different values or states.

#### World

The `World` Facet identifies the external referent required to interpret the Data.

For example:

```text
location = port
```

does not identify which port, at what time, under which jurisdiction, or in relation to which voyage.

The required referent may exist outside the artifact itself:

* the real vessel represented by an identifier;
* the geographic region to which a measurement applies;
* the population included in a dataset;
* the policy version governing a decision;
* the event being reported;
* the period during which a value is valid; or
* the authority under which an instruction should be followed.

Two participants can agree on a value's Meaning and Structure while applying it to different parts of the World.

#### The Complete Gauge

Interpretation therefore depends on more than Data alone:

$$
\text{Interpretation} = f(\text{Data}, \text{Meaning}, \text{Structure}, \text{World})
$$

The same Data can produce different interpretations when any Context Facet changes:

$$
D_A = D_B
$$

while:

$$
f(D_A, M_A, S_A, W_A) \ne f(D_B, M_B, S_B, W_B)
$$

The protocol measures each Facet separately because agreement along one dimension does not establish complete Context agreement.

```text
                 Context Gauge

                 ┌───────────┐
                 │  Meaning  │
                 └─────┬─────┘
                       │
┌───────────┐          │          ┌───────────┐
│ Structure │ ─────── Data ────── │   World   │
└───────────┘                     └───────────┘
```

#### What the Gauge Observes

At each Context Facet crossing, the gauge asks:

1. Has the Facet been declared?
2. Does it carry a resolving `cgvu:` referent?
3. Can the immediate sender's intent and the current reader's interpretation be compared?
4. If compared, do their referents produce `MATCH` or `NO MATCH`?

A Context Facet is:

* `DARK` when it has no resolving `cgvu:` referent; or
* `LIT` when it has a resolving `cgvu:` referent.

A `DARK` Facet may happen to be interpreted as the sender intended. However, the system lacks the reference needed to verify that relationship.

It is therefore operationally ignorant:

$$
\text{sender intent} \stackrel{?}{=} \text{reader interpretation}
$$

A `LIT` Facet makes the interpretation explicitly referable. When both sides of a crossing expose comparable referents, the gauge can produce:

* `MATCH` when the referents are the same; or
* `NO MATCH` when the referents differ.

The gauge does not measure the intrinsic meaning of a concept. It measures the contextual conditions under which concept-bearing `Data` is interpreted and used.

#### A Gauge, Not a Judge

The Context gauge does not determine which participant is correct.

It does not declare:

* which unit an organization should use;
* which schema should govern an API;
* which generator or validator should be authoritative;
* which population a model should represent; or
* which external policy should control an action.


Those are governance decisions.

> The gauge provides a coordinate frame and categorical comparison states. It does not assign a numerical distance between two non-matching referents. Any proposed Context distance or magnitude must declare its per-facet comparison function, weighting, aggregation rule, and decision scope.

The gauge provides the measurement infrastructure needed to make those decisions explicit and inspectable.

> **The protocol does not guarantee alignment. It makes alignment measurable across Meaning, Structure, and World.**

Without the gauge, a workflow may still transmit Data and produce decisions, just as a person can draw without a ruler. But every handoff relies on unaided assumptions about whether the participants remain aligned.

With the gauge, those assumptions become observable questions:

```text
Is the required Context present?
            │
            ▼
Is each Facet DARK or LIT?
            │
            ▼
Can intent and interpretation be compared?
            │
            ▼
Do they MATCH or produce NO MATCH?
            │
            ▼
Does the difference affect a decision?
```

This is the foundational capability introduced by Track 1:

> **A reusable gauge for measuring the contextual conditions under which Data can be interpreted across people, artifacts, systems, networks, and time.**

## **Knowledge Graphs vs. Context Graphs**

A Knowledge Graph can represent facts, relationships, schemas, provenance, and external entities relevant to interpretation. The Context Graph Protocol's operational purpose is to measure whether the `Meaning`, `Structure`, and `World` interpretations applied by an immediate sender and reader are comparable and aligned at a particular boundary.

Because the ways an intent and an interpretation can fail to be equivalent are open-ended, an experiment or operational system must declare which known unknowns it will observe. In the Context Graph Protocol, an `Intent Map` provides that declared scope: it identifies the relevant events, Context Facets, answer spaces, and routing rules. Its responses may be deterministic or may invoke inference. The events and measurements produced under an `Intent Map` may be collected and analyzed as _Context Intelligence_, which is outside the scope of this notebook.

| Dimension | Knowledge Graph | Context Graph Protocol |
| --- | --- | --- |
| Purpose | Represents what is known | Measures whether intent and interpretation align at a boundary |
| Unit | Entities and relationships | The Context Facet crossing: `DARK`/`LIT`, `MATCH`/`NO MATCH` |
| Where it sits | A curated resource that applications consult | At the interaction itself: an Observatron on a declared boundary |
| Who closes the loop | Knowledge engineers, through curation away from the interaction | The participants, in the moment: an `ASK` can be answered mid-interaction |
| When it operates | After definitions have been curated | Before an Ontology is operationalized — as early as a user's query |
| Relationship to the UI | A display or consultation surface, via custom application code | An addressable handler: the declared answer space becomes the UI element |
| URIs | Dereference to descriptions of entities | `cgvu:` addresses resolve by in-memory comparison — match, or no match — with nothing hosted at them |
| What it carries | Definitions | Rules of engagement — when to observe, what to `ASK`, what may be assumed: heuristics for coordinating understanding, with definitions remaining at the addresses the rules point to |

A key challenge is that each Context Facet is traditionally owned by a different community:

| Context Facet | Traditional custodian | Typical artifacts |
| --- | --- | --- |
| `Meaning` | Ontologists and taxonomists | Ontologies, controlled vocabularies, concept libraries |
| `Structure` | API and data engineering | API specs, schemas, formats, validators |
| `World` | The business domain: operations, subject-matter experts, policy owners — and the users themselves | Policy documents, master and reference data, situational knowledge — often tacit rather than curated |

> As a protocol requiring only addressable URIs and a declared registry of rules — an `Intent Map` — the Context Graph Protocol is designed to make all three sources of reference operationally available in **real-time, continuous communication**.

The Context Graph Protocol's role is deliberately narrow: to surface whether any such reference is actually in play at a crossing at all — `DARK` or `LIT` — and, when both sides expose comparable referents, whether they align — `MATCH` or `NO MATCH` — **between the immediate sender's intent and the reader's interpretation at a single crossing**.

# Introducing **`Dark Context`**

In the W3C Context Graph Community Group, we break down information we use for decisions into three **```Context Facets```** to reveal ambiguity and incomplete information in our understanding of data.

**Four Facets, three of which can go ```Dark```:**
- _```Data```_: What information is being examined?
- _**```Meaning```**_: What is the human-readable definition of the information?
- _**```Structure```**_: What are the machine-readable constraints on the representation?
- _**```World```**_: How does the data relate to the environment?

`Data` is present by observation. `Meaning`, `Structure`, and `World` are `DARK` when no resolving `URI` is carried at the observed crossing. A reader may still assign an interpretation, but fidelity to the immediate sender has not been established through resolving evidence.

A `Context Facet` is `LIT` when it carries a resolving `URI` and `DARK` when it does not. `LIT` does not automatically mean `MATCH`; the carried referent can still differ from the immediate sender’s referent.

Each `Context Facet` can completely change our understanding of the data — a shift in perspective at exactly the point where one system's assumptions meet another's.

## Minimal `Intent Map` Example

> An `Intent Map` is merely a list of instructions for what `intent` to `interpretation` events to observe, and what to do if those event conditions are met.


Minimal `Intent Map` example:

> _Rule:_ **Observe if the word `temperature` is used in a table column**

> _Action:_ **Verify the intent of `temperature` in the three `Context Facets`**

### `Intent Map` explanation:
- If the ```Meaning``` of "temperature" is Celsius, it's a warm day; if Fahrenheit, it's below fresh water's freezing point (32°F) and within half a degree of salt water's (~28.6°F).

- If the ```Structure``` of the data has a Schema that rounds the data up or down, then a recorded 29 could be anything from 28.5 to 29.5 — above or below the ~28.6°F threshold at which salt water freezes — and we don't know if a rounding rule was applied at all.

- If the ```World``` state the data is pointing to is Salt Water, then it may have a significantly different freezing temperature.


### Impact on Decision Making
Each of the three ```Context Facets``` can break our understanding, and each failure point has real economic costs.

### Operational Implementation
An Intent Map is something that evolves over time, that can be distributed at scale to business units, applications, teams, and individuals, that everyone can benefit from.

_For example:_ If a new instruction is added to an `Intent Map` then everyone can gain from a check on ambiguity. Likewise, if everyone on a team exclusively points to _Celsius_, then a system can transparently share _Celsius_ is assumed without needing to ask every time, showing users which assumptions are in play.


## **Core Components**

### ```Observatron```
A **frameless evaluator** anchored at a declared boundary around data. It observes what crosses, mints ```Spike```s, counts what is present and what is blank, and routes the measured state to the ```Governor```. **Normative:** *An Observatron holds no frame and does not choose or bind an interpretation.* That frame-freedom is exactly why one can be placed anywhere — over a chat input in a web page, on a backend file sent to an AI, on an API hop.

### ```Spike```
The four-facet object an Observatron mints when a piece of ```Data``` crosses its boundary: the Data itself plus its ```Meaning```, ```Structure```, and ```World``` facets. _For example, the column ```temperature``` crossing the boundary becomes a Spike._

### ```Intent Map```
The declared rulebook the Observatron reads: which events to observe, where, and — per facet — the **declared answer space** (the *shards*: the finite set of values the facet could take). Declaration is what makes darkness *countable*. When a rule matches, the Intent Map names the response, which may be deterministic or inferred. _For example, an Observatron can be instructed to ```ASK``` the user whether ```temperature``` means Celsius or Fahrenheit._

An Intent Map MUST be consumable by the Governor API.

### ```Governor```
A **decision model for routing**. Given the measured state of an Observatron's Spikes under the Intent Map, the Governor returns exactly one of ```ACT```, ```ASK```, or ```HALT``` — dispatch verdicts routed to addressable handlers, never semantics. ```ACT``` means "invoke the referenced action" (which may itself be a recorded no-op); the meaning of every action lives at its URI, not in the gate. ```HALT``` absorbs: once a halt condition fires, no other input lowers the response.

### ```Certificate```
A minimal set of *answers* that locks the verdict regardless of every remaining
Dark facet — in formal logic, the prime implicants of the decision. Reaching a
certificate **neutralizes** the rest: still Dark, no longer able to react.

### ```Actor```
A human, an LLM, and a system (an API, a batch job) are **the same kind of participant**: anything that reads through a frame and either records an intent or flags a condition. No definition, rule, or theorem in this protocol may branch on which. Sender and receiver are *roles in an event*, not kinds of thing. An ```Actor``` creates actions (events).

### ```Interaction at the Boundary```
An event has occurred on something the ```Observatron``` is observing, this is where feedback loops occur, and the state that the ```Observatron``` monitors with an ```Intent Map```. An interaction is either ```input```, ```process```, or ```output```.

The Context Graph Protocol is a minimal, **common gauge** of Dark Uncertainty — read by Observatrons, declared in Intent Maps, routed by Governors — across all technology layers and organizational networks.


**They meet in the Intent Map.** A rule addressable in `cgp` names *when* to observe and *what* to declare; the candidate answer spaces it declares are `cgvu:` addresses (`cgvu:/temperature/meaning/celsius`) that facets bind to. Hosted machinery on the web, virtual gauge in memory: **`cgp` hosts, `cgvu:` gauges.**

### **Decision Outputs and Routing Responsibilities**

| Component or Concept | Responsibility or Output |
|---|---|
| `Observatron` | The top-level boundary object containing the cooperating components. |
| `Context State` | Holds the boundary events, `Spike`s, facets, values, and statuses available for evaluation. |
| `Intent Map` | Declares what to observe, the applicable answer spaces and conditions, and the available routing instructions. |
| `Governor` | Evaluates the Context State, Intent Map, metadata, and applicable policies. It produces a Governor `Verdict` and selects a `Decision Route`. |
| Governor `Verdict` | The routing decision output: `ACT`, `ASK`, or `HALT`. |
| `Decision Route` | The URI selected by the Governor to identify the behavior to perform. |
| `Handler Registry` | Binds a `Decision Route` URI to the registered `Handler` responsible for that route. |
| `Handler` | Performs the route’s behavior and emits events through which state may subsequently change. |
| Operational `Verdict` | The separate output of an `Operational Decision`, such as the captain’s `GO` or `NO-GO`. |

A `Verdict` is always the output of a particular decision model. Therefore, the decision model should be named whenever the term could be ambiguous:

| Decision Model | Example `Verdict`s |
|---|---|
| Governor routing decision | `ACT`, `ASK`, `HALT` |
| Captain’s Operational Decision | `GO`, `NO-GO` |

A `Decision Route` URI identifies behavior; it does not necessarily indicate browser navigation. Its registered `Handler` may open a feedback loop, request clarification, display a notification, invoke an action, run a command chain, record a no-op, halt processing, or navigate to another resource.

## Decision Key Concepts


*`Outer Decision Perspective`*: What is the operational system deciding?

> `Decision System` → `Operational Decision` → `Verdict`

*`Decision Shape`*: Which `Context Facet`s are essential to a particular `Operational Decision`, and which are inert?

*`Inner Decision Perspective`*: What must the Context Graph do so that the decision can be made safely?

> `Governor`: `Context State` + `Intent Map` → `Decidability Gate` → `Decision Route`


## Outer Decision Perspective
The perspective on the operational system whose decisions the Context Graph exists to support, viewed from outside the Context Graph machinery.

```text
Decision System
├── Decision Rule(s)
└── Operational Decision
        └── Verdict
```

| **Decision System**      | The bounded system in which interpreted information is evaluated under one or more `Decision Rule`s.            |
| **Decision Rule**        | The logic that maps interpreted information to a `Verdict`: `if this → do that`.                                |
| **Operator**             | A human `Actor` who operates, configures, invokes, or supplies intent to a `Decision System`.                   |
| **Operational Decision** | A particular evaluation of a `Decision System` in a particular state under its applicable `Decision Rule`(s).  |
| **Verdict**              | The outcome of an `Operational Decision`, such as `GO`, `NO-GO`, approve, or reject.                            |



## Decision Shape Concepts

**`Decision Shape`** — The set of `Essential Facet`s for a particular `Operational Decision`.

| Term                | Definition                                                                                                                                                |
| ------------------- | --------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **Decision Shape**  | The set of `Essential Facet`s for a particular `Operational Decision`.                                                                                    |
| **Essential Facet** | A `Context Facet` whose interpretation can change the `Verdict` of a particular `Operational Decision`.                                                   |
| **Inert Facet**     | A `Context Facet` whose interpretation cannot change the `Verdict` of a particular `Operational Decision` and is therefore outside that `Decision Shape`. |

```text
All Context Facets for an Operational Decision
                    │
          ┌─────────┴─────────┐
          ▼                   ▼

          │                   │
          ▼                   ▼

```

The perspective from within the Context Graph machinery on how the context required by an `Operational Decision` is measured, resolved, and routed.

| **Decidability Gate** | The Governor's evaluation of whether the current context fixes the `Verdict` of an `Operational Decision` regardless of any remaining unresolved facets.                                                         |
| **Decision Route**    | The route selected by the `Governor`, such as `ACT`, `ASK`, or `HALT`, and any addressable handler associated with that route.                                                                                   |

# **Notebook Tracks**

This notebook investigates Context Graph behavior through four distinct experimental tracks. Each track examines a different boundary or level of analysis while using the same core vocabulary of Data, Context Facets, Observatrons, Intent Maps, Governors, and `ACT` / `ASK` / `HALT` routing.

## **Track 1 — The Context Loop**

Track 1 follows information through a complete communication loop across four declared artifacts:

* `table-1`
* `table-2`
* `query-1`
* `response-1`

crossed by six declared reading acts between participants.

Its standard unit is the **Context Facet crossing**: one reading act meeting one Context Facet.

Track 1 asks:

> Where can Dark Context be misread as information crosses boundaries, how can those misreadings propagate, and what can they cost?

The loop contains 54 Context Facet crossings: six reading acts, each meeting nine declared facet places. The track:

1. counts Context Facet places and crossings;
2. measures each crossing as `DARK` or `LIT`;
3. compares immediate sender intent with reader interpretation;
4. demonstrates how local fidelity can preserve an earlier misinterpretation;
5. prices selected decision-changing interpretations through Verdict mismatches, cash outcomes, and regret; and
6. measures the effect of resolving Context Facets through routed `ASK` interactions.

## **Track 2 — Decision Shapes**

Track 2 narrows the scope to one artifact, `table-1`, under one Operational Decision: the captain’s `GO` or `NO-GO` Verdict.

Its standard unit is the **candidate interpretation state**: one complete assignment to the candidate Context Facets declared for that decision.

Track 2 asks:

> Which declared interpretations can change this particular Verdict, and which cannot?

The experiment evaluates all 2^6 = 64 complete assignments to six candidate facets. It certifies each candidate as either:

* `Essential` — changing the facet can change the Verdict; or
* `Inert` — changing the facet cannot change the Verdict.

The resulting Decision Shape is proven to be a minimal four-facet function: four candidates are essential, two are inert, and the 64 complete states collapse into 16 decision-equivalent classes.

## **Track 3 — Context-Aware Join Proof**

Track 3 moves the Context Gauge to a data-integration boundary between `table-1` and `table-2`.

Its standard unit is the **join-key interpretation state and resulting record pairing**.

Track 3 asks:

> Can a join execute successfully while pairing the wrong evidence, and can declared Context make the correct pairing measurable?

The experiment compares:

* a context-blind join performed on raw values;
* a Governor evaluating the Context Facets of the join key; and
* a resolved join performed after declared canonicalization.

It measures:

* `MATCH`, `NO MATCH`, `DARK`, and transformable join conditions;
* Governor routing through `ACT`, `ASK`, and `HALT`;
* pair precision and recall;
* which evidence enters the downstream decision;
* the resulting Operational Verdict; and
* economic regret under the declared payoff ledger.

## **Track 4 — Context Networks and Feedback**

Track 4 expands the experiment from one Observatron to a network of Observatrons coordinated through a Decidatron, shared registries, and shared resolution records.

Its units include the **member-level Context Facet resolution**, the **network resolution record**, and the **set of Context Facets made LIT**.

Track 4 asks:

> How can a network share resolutions without silently imposing interpretations, learn which questions are most valuable, and amortize the cost of asking them?

The track demonstrates:

1. network-wide vocabulary and registry checks;
2. shared resolution records with member and cohort provenance;
3. cohort suggestions that remain `DARK` until the affected member confirms;
4. the cost of silently auto-adopting a majority interpretation;
5. `ASK` as a negotiation route capable of minting a new scoped referent;
6. exact Hartley uncertainty across the declared answer spaces;
7. an expected-regret lattice over every LIT subset;
8. exhaustive ranking of all possible question orders; and
9. amortized disclosure, where one recorded resolution can serve every participating member.

## Virtual URI Addresses

### Two Address Spaces: `cgp:` and `cgvu:`

The notebook uses two address spaces with different resolution models:

* `cgp:` — **Context Graph Protocol address** — identifies a protocol resource or registered Handler behavior. A deployment’s registry binds the `cgp:` address to its implementation, which may be hosted at any HTTP or HTTPS endpoint chosen by the implementer.

* `cgvu:` — **Context Graph Virtual URI** — identifies a value, referent, or behavior resolved entirely within a declared local or in-memory registry. No hosted endpoint is required.

A `cgp:` address names the stable protocol route; it does not encode the route’s physical server location. For example:

```text
cgp:/track-3/handlers/canonicalize-then-join
```

A production registry might bind that address to:

```text
https://api.example.com/handlers/canonicalize-then-join
```

The endpoint may live anywhere the deployment permits. Changing that endpoint does not change the protocol-level `cgp:` address used by the Governor and Handler Registry.

A `cgvu:` address resolves locally. For example:

```text
cgvu:/temperature/meaning/celsius
```

Nothing needs to be hosted at that address. The applicable registry supplies the value, referent, or behavior associated with it.

In this notebook, both address spaces resolve through declared Python registries:

* `cgp:` addresses stand in for deployable protocol routes and dispatch to registered Handlers.
* `cgvu:` addresses support virtual, in-memory resolution when a hosted route is neither available nor required.

The notebook uses these prefixes as declared protocol conventions. It does not claim that either prefix is an IANA-registered URI scheme.

| Prefix  | Resolution model                                                                      | Used when                                                                      |
| ------- | ------------------------------------------------------------------------------------- | ------------------------------------------------------------------------------ |
| `cgp:`  | Registry binding to a protocol resource or Handler, optionally hosted over HTTP/HTTPS | The address represents deployable protocol behavior or another hosted resource |
| `cgvu:` | Resolution within a declared local or in-memory registry                              | No hosted endpoint is available or required                                    |

In either case, the address must be declared, resolve within the applicable environment, and support the required comparison or dispatch operation.

The shared utility functions below define how declared choices become `cgvu:` addresses and how a supplied choice space becomes a local registry. They do not declare an experimental track’s answer space or instantiate its registry.

> A facet is `LIT` when it carries a resolving address and `DARK` when it does not. The resolving address may use either convention when appropriate. A declared value such as `cgp:none` or `cgvu:none` may still be addressable and therefore `LIT`; darkness means the absence of a resolving address, not the presence of a value named “none.”


````python
### Shared `cgvu:` Address Utilities
import re


def slug(choice):
    """Convert one declared choice into a cgvu path segment."""
    return re.sub(
        r"[^a-z0-9]+",
        "-",
        choice.lower(),
    ).strip("-")


def candidate_uris(choice_space, data, facet):
    """Return the cgvu addresses declared by a supplied choice space."""
    return tuple(
        f"cgvu:/{data}/{facet}/{slug(choice)}"
        for choice in choice_space[data][facet]
    )


def uri_registry(choice_space):
    """Build a URI registry from an explicitly supplied choice space."""
    return [
        uri
        for data in choice_space
        for facet in choice_space[data]
        for uri in candidate_uris(
            choice_space,
            data,
            facet,
        )
    ]
````

## Track 1-A: A Precise Measure of Context in a Context Loop

Section 1-A introduces the `Observatron` as an instrument for measuring Context Facets across the complete Context Loop.

The loop contains six reading acts:

1. the data engineer sends `table-1` to the user;
2. the user sends `table-1` to the LLM;
3. the data engineer sends `table-2` to the user;
4. the user sends `table-2` to the LLM;
5. the user sends `query-1` to the LLM; and
6. the LLM sends `response-1` to the user.

For each observed artifact field, the `Observatron` mints a `Spike` containing the Data and its `Meaning`, `Structure`, and `World` Context Facets.

A **Context Facet place** is one Context Facet attached to one artifact-field Spike:

$$
\text{one Spike} \times \text{one Context Facet}
$$

A **Context Facet crossing** is one reading act in which an artifact carries that Context Facet across a boundary:

$$
\text{one reading act} \times \text{one Context Facet carried by the artifact}
$$

One Context Facet place can participate in more than one crossing because the same artifact can be read more than once.

Each observed Context Facet is:

- `DARK` when it carries no resolving `cgvu:` URI; or
- `LIT` when it carries a resolving `cgvu:` URI.

- 18 declared `cgvu:` addresses;

Section 1-A establishes measurement only. Comparison begins in Section 1-B, while decision consequences begin in Section 1-C.

### **Track 1 Intent Map**

Track 1 now converts its declared binary choice space into an executable `Intent Map`.

For each observed data-field name, the map declares the candidate `cgvu:` addresses for its `Meaning`, `Structure`, and `World` facets.

This Intent Map belongs only to Track 1.

````python
import json


def build_loop_intent_map(choice_space):
    """Build Track 1's phrase-detection Intent Map."""
    return {
        "rules": [
            {
                "if": {
                    "phrase-detected": data,
                },
                "then": [
                    {
                        "facet": facet,
                        "candidates": list(
                            candidate_uris(
                                choice_space,
                                data,
                                facet,
                            )
                        ),
                    }
                    for facet in choice_space[data]
                ],
            }
            for data in choice_space
        ],
    }


LOOP_INTENT_MAP = build_loop_intent_map(LOOP_CHOICES)

rule_count = len(LOOP_INTENT_MAP["rules"])
facet_declaration_count = sum(
    len(rule["then"])
    for rule in LOOP_INTENT_MAP["rules"]
)

assert rule_count == 3
assert facet_declaration_count == 9

print(json.dumps(LOOP_INTENT_MAP, indent=2))

print(
    f"\nTrack 1 Intent Map: {rule_count} rules, "
    f"{facet_declaration_count} facet declarations"
)
````

### **Observatron Class**

````python
class Observatron:
    """Track 1 boundary container instantiated with an Intent Map.

    One Spike is minted per matching observed column. The Observatron holds
    its Spikes and their Context Facet values.
    """

    def __init__(self, name, intent_map):
        self.name = name
        self.intent_map = intent_map
        self.rules = intent_map["rules"]
        self.spikes = []
        self.facets = {}

    def detect(self, text):
        """Return the Track 1 rules whose phrases match the observed text."""
        return [
            rule
            for rule in self.rules
            if rule["if"]["phrase-detected"] in text.lower()
        ]

    def observe(self, artifact, columns):
        """Mint Spikes for matching columns and initialize their facets DARK."""
        minted = []

        for column in columns:
            for rule in self.detect(column):
                spike = {
                    "spike": len(self.spikes) + 1,
                    "artifact": artifact,
                    "data": column,
                    "rule": rule["if"]["phrase-detected"],
                }

                self.spikes.append(spike)

                for facet in LOOP_CONTEXT_FACETS:
                    self.facets[(spike["spike"], facet)] = ""

                minted.append(spike)

        return minted

    def uri(self, spike, facet):
        """Return the facet's resolving cgvu address, or an empty string."""
        return self.facets[(spike, facet)]

    def state(self, spike, facet):
        """Return DARK when unresolved and LIT when an address is present."""
        return (
            "DARK"
            if self.facets[(spike, facet)] == ""
            else "LIT"
        )

    def table(self):
        """Return the Observatron's current Spike and facet state."""
        return {
            "spike": [
                spike["spike"]
                for spike in self.spikes
            ],
            "artifact": [
                spike["artifact"]
                for spike in self.spikes
            ],
            "data": [
                spike["data"]
                for spike in self.spikes
            ],
            **{
                facet: [
                    self.state(spike["spike"], facet)
                    for spike in self.spikes
                ]
                for facet in LOOP_CONTEXT_FACETS
            },
        }

    def counts(self):
        """Count the Context Facets currently held by this Observatron."""
        dark = sum(
            1
            for uri in self.facets.values()
            if uri == ""
        )

        return {
            "spikes": len(self.spikes),
            "context_facets": len(self.facets),
            "dark": dark,
            "lit": len(self.facets) - dark,
        }
````

### **Facet Crossings Calculator**

````python
def crossings():
    """Return one row per Context Facet crossing.

    A crossing is one reading act in which an artifact carries one
    Context Facet across a boundary. Every Spike carried by the artifact
    is exposed at every reading act involving that artifact.
    """
    rows = {
        "index": [],
        "reading act": [],
        "from": [],
        "to": [],
        "artifact": [],
        "spike": [],
        "data": [],
        "facet": [],
    }

    n = 0

    for act, (source, reader, artifact) in enumerate(
        LOOP_READING_ACTS,
        1,
    ):
        for spike in observatron_1.spikes:
            if spike["artifact"] != artifact:
                continue

            for facet in LOOP_CONTEXT_FACETS:
                n += 1
                rows["index"].append(n)
                rows["reading act"].append(act)
                rows["from"].append(source)
                rows["to"].append(reader)
                rows["artifact"].append(artifact)
                rows["spike"].append(spike["spike"])
                rows["data"].append(spike["data"])
                rows["facet"].append(facet)

    return rows


X = crossings()

assert len(X["index"]) == 54

print_table(X)
print(f"total crossings: {len(X['index'])}")
````

## **Track 1-B: Comparing Intent with Interpretation at Context Facet Crossings**

Track 1-B asks:

> Where do immediate sender intent and current reader interpretation `MATCH` or produce `NO MATCH`, and what happens to fidelity across consecutive crossings?

An **intent** is the interpretation an immediate sender means to communicate.

An **interpretation** is the referent assigned by the current reader.

A `cgvu:` URI identifies a declared referent. The referent records what an intent or interpretation points to; it is not itself the cognitive act of intending or interpreting.

For this experiment, one source intent is recorded when each artifact-field Context Facet is authored. That recorded source intent remains fixed as the experiment’s source of truth.

When a participant reads an artifact and later sends it onward, the interpretation they currently hold becomes their intent for the next handoff.

At each measured crossing:

- `MATCH` means the immediate sender’s intent and current reader’s interpretation point to the same referent.
- `NO MATCH` means they point to different referents.

`MATCH` and `NO MATCH` are comparison results. They are not Context states, Governor Verdicts, or Operational Verdicts.

The 12 artifact-field Spikes each contain three Context Facets:

$$
12 \times 3 = 36\ \text{recorded source intents}
$$

The loop contains 54 Context Facet crossings:

$$
54\ \text{crossings}
=
54\ \text{reader interpretations}
=
54\ \text{comparison results}
$$

For the sampled calculations, every source-intent selection and every reader-interpretation selection at a `DARK` crossing is an independent, equally likely selection from the facet’s declared binary answer space.

These are controlled assumptions of the experiment. Randomness, independence, equal likelihood, and binary answer spaces are not inherent properties of the Context Graph Protocol.

### **Comparisons of Facet URI Matches in the Loop**

The table adds each crossing’s current `DARK` or `LIT` state and its two declared `cgvu:` candidates. This establishes the declared answer space within which the experiment records source intents and reader interpretations.

````python
def declared_referents(intent_map):
    """Flatten the Intent Map to (data field, facet) -> declared referents."""
    return {
        (
            rule["if"]["phrase-detected"],
            declaration["facet"],
        ): declaration["candidates"]
        for rule in intent_map["rules"]
        for declaration in rule["then"]
    }


def with_declared_referents(crossing_rows, obs, intent_map):
    """Append current bindings, states, and declared referents."""
    lookup = declared_referents(intent_map)
    rule_of = {
        spike["spike"]: spike["rule"]
        for spike in obs.spikes
    }

    out = {
        key: list(values)
        for key, values in crossing_rows.items()
    }

    out["resolving uri"] = []
    out["context state"] = []
    out["declared referent 1"] = []
    out["declared referent 2"] = []

    for spike, facet in zip(
        crossing_rows["spike"],
        crossing_rows["facet"],
    ):
        first, second = lookup[
            (rule_of[spike], facet)
        ]

        out["resolving uri"].append(
            obs.uri(spike, facet)
        )
        out["context state"].append(
            obs.state(spike, facet)
        )
        out["declared referent 1"].append(first)
        out["declared referent 2"].append(second)

    return out


XC = with_declared_referents(
    X,
    observatron_1,
    LOOP_INTENT_MAP,
)

print_table(XC)

facet_places = {
    (spike, facet): state
    for spike, facet, state in zip(
        XC["spike"],
        XC["facet"],
        XC["context state"],
    )
}

dark_places = sum(
    state == "DARK"
    for state in facet_places.values()
)

dark_crossings = XC["context state"].count(
    "DARK"
)

assert len(facet_places) == 36
assert dark_places == 36
assert len(XC["index"]) == 54
assert dark_crossings == 54

print_table({
    "Context Facet places": [len(facet_places)],
    "LIT": [len(facet_places) - dark_places],
    "DARK": [dark_places],
    "% DARK": [
        f"{dark_places / len(facet_places):.0%}"
    ],
})

print_table({
    "Context Facet crossings": [len(XC["index"])],
    "LIT": [
        len(XC["index"]) - dark_crossings
    ],
    "DARK": [dark_crossings],
    "% DARK": [
        f"{dark_crossings / len(XC['index']):.0%}"
    ],
})
````

### **Exact Probability That All Remaining Dark Crossings Produce `MATCH`**

With 54 independent `DARK` crossings and two equally likely interpretations per crossing, this cell calculates the exact probability that every current reader selects the immediate sender’s referent:

$$
P(\text{all crossings MATCH})
=
\left(\frac{1}{2}\right)^{54}
$$

This is an exact calculation under the declared experiment. It is not a sampled result, an organizational error-rate estimate, or a property of the protocol.

````python
from fractions import Fraction
from decimal import Decimal, getcontext


dark_crossing_count = XC[
    "context state"
].count("DARK")

answer_space_sizes = {
    len(referents)
    for referents in declared_referents(
        LOOP_INTENT_MAP
    ).values()
}

assert answer_space_sizes == {2}

all_crossings_match_probability = (
    Fraction(1, 2) ** dark_crossing_count
)

assert (
    all_crossings_match_probability
    == Fraction(1, 2**54)
)

getcontext().prec = 40

decimal_probability = (
    Decimal(
        all_crossings_match_probability.numerator
    )
    / Decimal(
        all_crossings_match_probability.denominator
    )
)

print_table({
    "Context Facet crossings": [
        len(XC["index"])
    ],
    "DARK crossings": [
        dark_crossing_count
    ],
    "Referents per crossing": [2],
    "Exact P(all crossings MATCH)": [
        f"{all_crossings_match_probability.numerator}/"
        f"{all_crossings_match_probability.denominator}"
    ],
    "Decimal P(all crossings MATCH)": [
        f"{decimal_probability:.40f}"
    ],
    "Exact odds": [
        "1 in "
        f"{all_crossings_match_probability.denominator:,}"
    ],
})
````

### **Sensitivity Curve as Crossings Are Removed from Uncertainty**

The preceding calculation gives the exact probability that all 54 current reader interpretations match their immediate senders’ intents.

For this sensitivity curve, a crossing is removed from uncertainty only when the reader is bound to the referent intended by the immediate sender. Merely making a crossing `LIT` does not satisfy that condition if the carried referent differs from the sender’s referent.

If $d$ independent binary crossings remain `DARK`, then:

$$
P(\text{all remaining crossings MATCH})
=
\left(\frac{1}{2}\right)^d
$$

Each row shows the exact probability for one possible number of remaining `DARK` crossings.

This is a mathematical sensitivity calculation under the experiment’s independent, equally likely binary assumptions. It is not a sampled result, an organizational error-rate estimate, or a claim that every `LIT` referent necessarily matches the sender’s intent.

The calculation measures local fidelity at each crossing. It does not establish fidelity to the original source intent.

````python
def dark_crossing_curve(
    initial_dark_crossings,
    width=40,
):
    rows = {
        "Crossings removed from uncertainty": [],
        "DARK crossings remaining": [],
        "Exact P(all remaining MATCH)": [],
        "Odds all remaining MATCH": [],
        "Probability gauge": [],
    }

    for dark_remaining in range(
        initial_dark_crossings,
        -1,
        -1,
    ):
        probability = (
            Fraction(1, 2) ** dark_remaining
        )

        rows[
            "Crossings removed from uncertainty"
        ].append(
            initial_dark_crossings
            - dark_remaining
        )

        rows[
            "DARK crossings remaining"
        ].append(
            dark_remaining
        )

        rows[
            "Exact P(all remaining MATCH)"
        ].append(
            f"{probability.numerator}/"
            f"{probability.denominator}"
        )

        rows[
            "Odds all remaining MATCH"
        ].append(
            "certain"
            if probability == 1
            else f"1 in {probability.denominator:,}"
        )

        rows["Probability gauge"].append(
            "█" * round(
                float(probability) * width
            )
            or "·"
        )

    return rows


DARK_CROSSING_CURVE = dark_crossing_curve(
    LOOP_DARK_CROSSING_COUNT
)

assert (
    DARK_CROSSING_CURVE[
        "Exact P(all remaining MATCH)"
    ][0]
    == f"1/{2**54}"
)

assert (
    DARK_CROSSING_CURVE[
        "Exact P(all remaining MATCH)"
    ][-1]
    == "1/1"
)

print_table(DARK_CROSSING_CURVE)
````

### **Reading the Sampled Experiment Summary**

The experiment evaluates 54 Context Facet crossings. At every crossing, it compares:

- the intent held by the immediate sender; and
- the interpretation assigned by the reader.

Each crossing produces one comparison result:

- `MATCH` means the reader’s interpretation points to the same `cgvu:` referent as the sender’s intent.
- `NO MATCH` means the two point to different referents.


#### **Local Crossing Match Rate**

_`Local crossing match rate`_: In one measured run, the percentage of Context Facet crossings where the reader’s current interpretation matches the immediate sender’s intent.

> The `local crossing match rate` is a local fidelity measure. It answers: “In this run, how often did a reader interpret a Context Facet the same way the immediate sender intended it?”

The **local crossing match rate** is the proportion of crossing comparisons that produce `MATCH`:

$$
\text{local crossing match rate}
=
\frac{\text{number of MATCH results}}
{\text{number of Context Facet crossings}}
$$

For example, if the experiment reports:

```text
26 MATCH + 28 NO MATCH = 54 crossing comparisons
```

then:

$$
\text{local crossing match rate}
=
\frac{26}{54}
\approx
48\%
$$

This means that, in this sampled run, the reader’s interpretation matched the sender’s intent at 26 of the 54 crossings.

It does **not** mean that 26 interpretations matched the original source intent. At a later handoff, a sender may communicate an interpretation acquired from an earlier reading. A local `MATCH` can therefore faithfully preserve an earlier misinterpretation.

The local crossing match rate tells us how often individual handoffs preserve the interpretation held by their immediate sender. It does not establish:

- whether that interpretation matches the original source intent;
- whether either interpretation is correct outside the experiment;
- whether a `NO MATCH` changes an Operational Verdict; or
- whether the resulting difference creates economic regret.

Those questions require the later propagation and Operational Decision analyses.

### **What “Draw” Means**

A **draw** is one pseudorandom selection from a Context Facet’s declared answer space.

For example, if `temperature.meaning` declares:

```text
Celsius
Fahrenheit
```

then pseudorandomly selecting either referent once is one draw.

The current experiment makes:

- 36 draws to establish the source intents recorded at authorship; and
- 54 draws to establish the reader interpretations assigned at crossings.

That produces 90 draws, but only 54 comparisons:

$$
36\ \text{source-intent draws}
+
54\ \text{reader-interpretation draws}
=
90\ \text{draws}
$$

$$
54\ \text{crossings}
=
54\ \text{intent-to-interpretation comparisons}
$$

A draw is therefore neither a crossing nor a comparison. It is only the experiment’s mechanism for selecting one declared referent.

Because “draw” describes an implementation detail rather than a protocol concept, the reader-facing output below reports source intents, reader interpretations, and crossing comparisons directly.

````python
import random


def artifact_authors(reading_acts):
    """Return the sender at each artifact's first appearance."""
    authors_by_artifact = {}

    for sender, _reader, artifact in reading_acts:
        authors_by_artifact.setdefault(
            artifact,
            sender,
        )

    return authors_by_artifact


ARTIFACT_AUTHOR = artifact_authors(
    LOOP_READING_ACTS
)

# Unseeded by design: each run draws a fresh sample from the declared answer
# spaces, so the MATCH count varies run to run. The seeded replay cell below
# (SEED = 4137958253) is the exact-reproducibility witness.
RNG = random.Random()

MATCH = "MATCH"
NO_MATCH = "NO MATCH"

COMPARISON_COLOUR = {
    MATCH: "#1a7f37",
    NO_MATCH: "#d1242f",
}


def painted_comparisons(rows):
    """Color comparison results for display without changing the data."""
    return {
        **rows,
        "comparison result": [
            Raw(
                f'<span style="color:'
                f'{COMPARISON_COLOUR[result]};'
                f'font-weight:600">'
                f'{result}</span>'
            )
            for result in rows[
                "comparison result"
            ]
        ],
    }


class CrossingExperiment:
    """Record intent and interpretation at each measured crossing.

    One source intent is selected once for each artifact-field
    Context Facet at authorship.

    At a DARK crossing, the current reader selects one referent
    from the declared answer space.

    If that participant sends the artifact onward, the
    interpretation they currently hold becomes their intent for
    the next handoff.
    """

    def __init__(
        self,
        name,
        obs,
        intent_map,
    ):
        self.name = name
        self.obs = obs
        self.rng = RNG

        self.lookup = declared_referents(
            intent_map
        )

        self.rule_of = {
            spike["spike"]: spike["rule"]
            for spike in obs.spikes
        }

        self.held_referents = {}
        self.source_intent_count = 0
        self.reader_interpretation_count = 0

    def candidates(
        self,
        spike,
        facet,
    ):
        """Return the declared answer space for one Context Facet."""
        return self.lookup[
            (self.rule_of[spike], facet)
        ]

    def record_source_intent(
        self,
        artifact,
        author,
        spike,
        facet,
    ):
        """Select and retain the source intent recorded at authorship."""
        key = (
            author,
            spike,
            facet,
        )

        if key not in self.held_referents:
            self.held_referents[key] = (
                self.rng.choice(
                    self.candidates(
                        spike,
                        facet,
                    )
                )
            )

            self.source_intent_count += 1

        return self.held_referents[key]

    def sender_intent(
        self,
        artifact,
        sender,
        spike,
        facet,
    ):
        """Return the referent the immediate sender currently means to send."""
        key = (
            sender,
            spike,
            facet,
        )

        if key in self.held_referents:
            return self.held_referents[key]

        assert (
            sender
            == ARTIFACT_AUTHOR[artifact]
        )

        return self.record_source_intent(
            artifact,
            sender,
            spike,
            facet,
        )

    def interpret(
        self,
        reader,
        spike,
        facet,
    ):
        """Select the current reader's referent at one DARK crossing."""
        referent = self.rng.choice(
            self.candidates(
                spike,
                facet,
            )
        )

        self.held_referents[
            (reader, spike, facet)
        ] = referent

        self.reader_interpretation_count += 1

        return referent

    def run(
        self,
        crossing_rows,
    ):
        """Return one complete audit row per measured crossing."""
        out = {
            key: []
            for key in (
                "index",
                "reading act",
                "from",
                "to",
                "artifact",
                "sender intent referent",
                "reader interpretation referent",
                "comparison result",
            )
        }

        columns = zip(
            crossing_rows["index"],
            crossing_rows["reading act"],
            crossing_rows["from"],
            crossing_rows["to"],
            crossing_rows["artifact"],
            crossing_rows["spike"],
            crossing_rows["facet"],
        )

        for (
            index,
            act,
            sender,
            reader,
            artifact,
            spike,
            facet,
        ) in columns:
            intent_referent = (
                self.sender_intent(
                    artifact,
                    sender,
                    spike,
                    facet,
                )
            )

            interpretation_referent = (
                self.interpret(
                    reader,
                    spike,
                    facet,
                )
            )

            result = (
                MATCH
                if intent_referent
                == interpretation_referent
                else NO_MATCH
            )

            out["index"].append(index)
            out["reading act"].append(act)
            out["from"].append(sender)
            out["to"].append(reader)
            out["artifact"].append(artifact)

            out[
                "sender intent referent"
            ].append(
                intent_referent
            )

            out[
                "reader interpretation referent"
            ].append(
                interpretation_referent
            )

            out[
                "comparison result"
            ].append(
                result
            )

        return out


experiment_1 = CrossingExperiment(
    "experiment-1",
    observatron_1,
    LOOP_INTENT_MAP,
)

EXPERIMENT_1_TABLE = experiment_1.run(
    X
)

print_table(
    painted_comparisons(
        EXPERIMENT_1_TABLE
    )
)

for (
    index,
    spike,
    facet,
    intent_referent,
    interpretation_referent,
) in zip(
    EXPERIMENT_1_TABLE["index"],
    X["spike"],
    X["facet"],
    EXPERIMENT_1_TABLE[
        "sender intent referent"
    ],
    EXPERIMENT_1_TABLE[
        "reader interpretation referent"
    ],
):
    answer_space = experiment_1.candidates(
        spike,
        facet,
    )

    assert intent_referent in answer_space

    assert (
        interpretation_referent
        in answer_space
    ), f"row {index} off-space"


facet_place_count = len(
    set(
        zip(
            X["spike"],
            X["facet"],
        )
    )
)

assert (
    experiment_1.source_intent_count
    == facet_place_count
    == 36
)

assert (
    experiment_1.reader_interpretation_count
    == len(X["index"])
    == 54
)

match_count = EXPERIMENT_1_TABLE[
    "comparison result"
].count(MATCH)

no_match_count = EXPERIMENT_1_TABLE[
    "comparison result"
].count(NO_MATCH)

assert (
    match_count
    + no_match_count
    == 54
)

assert all(
    result
    == (
        MATCH
        if intent_referent
        == interpretation_referent
        else NO_MATCH
    )
    for (
        intent_referent,
        interpretation_referent,
        result,
    ) in zip(
        EXPERIMENT_1_TABLE[
            "sender intent referent"
        ],
        EXPERIMENT_1_TABLE[
            "reader interpretation referent"
        ],
        EXPERIMENT_1_TABLE[
            "comparison result"
        ],
    )
)

local_crossing_match_rate = (
    match_count
    / len(
        EXPERIMENT_1_TABLE["index"]
    )
)

print(
    f"{experiment_1.name}: "
    "54 Context Facet crossing comparisons"
)

print(
    f"{experiment_1.source_intent_count} "
    "source intents recorded at authorship"
)

print(
    f"{experiment_1.reader_interpretation_count} "
    "reader interpretations sampled at crossings"
)

print(
    f"{match_count} MATCH + "
    f"{no_match_count} NO MATCH "
    "= 54 crossing comparisons"
)

print(
    "local crossing match rate: "
    f"{match_count}/54 "
    f"= {local_crossing_match_rate:.0%}"
)

print(
    "A local MATCH compares the current reader "
    "with the immediate sender; it does not prove "
    "fidelity to the original source intent."
)
````

### **When Interpretation Becomes Intent: How Misinterpretations Propagate**

When a reader later sends the artifact onward, their current interpretation becomes their intent for the next handoff.

The following proof enumerates all eight binary histories across two handoffs and compares:

1. the original source intent with the first reader’s interpretation;
2. the first reader’s interpretation, now communicated as intent, with the second reader’s interpretation; and
3. the final interpretation with the original source intent.

A local `MATCH` establishes fidelity to the immediate sender at one crossing. It does not establish fidelity to the original source intent.

The proof distinguishes two different probabilities:

- **Unconditional:** a randomly selected complete two-handoff history contains a first `NO MATCH` that is then faithfully propagated in $2/8=25\%$ of all histories.
- **Conditional:** given that the first handoff has already produced `NO MATCH`, faithful propagation at the second handoff occurs in $2/4=50\%$ of the eligible histories.

The proof also identifies endpoint restoration after two local `NO MATCH` results:

```text
source intent: A
first reader: B           → NO MATCH

B becomes the next sender’s intent

second reader: A          → NO MATCH
final vs original source  → MATCH
```

The endpoint agrees with the original source even though neither local handoff preserved its immediate sender’s intent.

````python
from itertools import product


PROPAGATION_DATA = "temperature"
PROPAGATION_FACET = "meaning"

PROPAGATION_REFERENTS = candidate_uris(
    LOOP_CHOICES,
    PROPAGATION_DATA,
    PROPAGATION_FACET,
)

assert len(PROPAGATION_REFERENTS) == 2

PROPAGATION_LABELS = {
    PROPAGATION_REFERENTS[0]: (
        f"A — {PROPAGATION_REFERENTS[0]}"
    ),
    PROPAGATION_REFERENTS[1]: (
        f"B — {PROPAGATION_REFERENTS[1]}"
    ),
}


def comparison_result(
    intent,
    interpretation,
):
    """Compare the declared referents of an intent and interpretation."""
    return (
        MATCH
        if intent == interpretation
        else NO_MATCH
    )


PROPAGATION_STATES = tuple(
    product(
        PROPAGATION_REFERENTS,
        repeat=3,
    )
)

assert len(PROPAGATION_STATES) == 8
assert len(set(PROPAGATION_STATES)) == 8

PROPAGATION_TABLE = {
    "state": [],
    "original source intent": [],
    "first reader interpretation": [],
    "handoff 1": [],
    "next sender intent": [],
    "second reader interpretation": [],
    "handoff 2": [],
    "final vs original source": [],
    "earlier misinterpretation faithfully propagated?": [],
}

for state_number, (
    source_intent,
    first_interpretation,
    second_interpretation,
) in enumerate(
    PROPAGATION_STATES,
    1,
):
    next_sender_intent = (
        first_interpretation
    )

    handoff_1 = comparison_result(
        source_intent,
        first_interpretation,
    )

    handoff_2 = comparison_result(
        next_sender_intent,
        second_interpretation,
    )

    final_vs_source = comparison_result(
        source_intent,
        second_interpretation,
    )

    faithfully_propagated = (
        handoff_1 == NO_MATCH
        and handoff_2 == MATCH
        and final_vs_source == NO_MATCH
    )

    PROPAGATION_TABLE[
        "state"
    ].append(
        state_number
    )

    PROPAGATION_TABLE[
        "original source intent"
    ].append(
        PROPAGATION_LABELS[
            source_intent
        ]
    )

    PROPAGATION_TABLE[
        "first reader interpretation"
    ].append(
        PROPAGATION_LABELS[
            first_interpretation
        ]
    )

    PROPAGATION_TABLE[
        "handoff 1"
    ].append(
        handoff_1
    )

    PROPAGATION_TABLE[
        "next sender intent"
    ].append(
        PROPAGATION_LABELS[
            next_sender_intent
        ]
    )

    PROPAGATION_TABLE[
        "second reader interpretation"
    ].append(
        PROPAGATION_LABELS[
            second_interpretation
        ]
    )

    PROPAGATION_TABLE[
        "handoff 2"
    ].append(
        handoff_2
    )

    PROPAGATION_TABLE[
        "final vs original source"
    ].append(
        final_vs_source
    )

    PROPAGATION_TABLE[
        "earlier misinterpretation "
        "faithfully propagated?"
    ].append(
        "YES"
        if faithfully_propagated
        else "NO"
    )


print(
    "Complete table: all 8 binary histories "
    "across two handoffs, not a sample."
)

print_table(
    PROPAGATION_TABLE
)


both_handoffs_match = sum(
    first == MATCH
    and second == MATCH
    for first, second in zip(
        PROPAGATION_TABLE[
            "handoff 1"
        ],
        PROPAGATION_TABLE[
            "handoff 2"
        ],
    )
)

endpoint_matches_source = (
    PROPAGATION_TABLE[
        "final vs original source"
    ].count(MATCH)
)

faithfully_propagated_count = (
    PROPAGATION_TABLE[
        "earlier misinterpretation "
        "faithfully propagated?"
    ].count("YES")
)

restored_after_two_no_match = sum(
    first == NO_MATCH
    and second == NO_MATCH
    and endpoint == MATCH
    for first, second, endpoint in zip(
        PROPAGATION_TABLE[
            "handoff 1"
        ],
        PROPAGATION_TABLE[
            "handoff 2"
        ],
        PROPAGATION_TABLE[
            "final vs original source"
        ],
    )
)

first_handoff_no_match = (
    PROPAGATION_TABLE[
        "handoff 1"
    ].count(NO_MATCH)
)

assert both_handoffs_match == 2
assert endpoint_matches_source == 4
assert faithfully_propagated_count == 2
assert restored_after_two_no_match == 2
assert first_handoff_no_match == 4

assert (
    Fraction(
        faithfully_propagated_count,
        first_handoff_no_match,
    )
    == Fraction(1, 2)
)

print_table({
    "Unconditional condition": [
        "Both handoffs MATCH",
        "Final interpretation matches original source",
        "First NO MATCH is faithfully propagated",
        "Endpoint restored after two local NO MATCH results",
    ],
    "Histories": [
        f"{both_handoffs_match}/8",
        f"{endpoint_matches_source}/8",
        f"{faithfully_propagated_count}/8",
        f"{restored_after_two_no_match}/8",
    ],
    "Unconditional probability": [
        "25%",
        "50%",
        "25%",
        "25%",
    ],
})

print_table({
    "Conditional question": [
        "Given handoff 1 is already NO MATCH, "
        "handoff 2 faithfully propagates it"
    ],
    "Eligible histories": [
        f"{faithfully_propagated_count}/"
        f"{first_handoff_no_match}"
    ],
    "Conditional probability": [
        "50%"
    ],
})
````

### **Dark Uncertainty: Operational Ignorance Across Context Facet Crossings**

A `DARK` Context Facet carries no resolving `cgvu:` URI at the observed crossing.

It does not prove that a mismatch occurred. It means that fidelity has not been established through resolving evidence.

> **Operational ignorance exists at a Context Facet crossing when the system lacks resolving evidence sufficient to determine whether the current reader’s interpretation matches the immediate sender’s intent.**

The test harness can record source-intent and reader-interpretation referents for audit while the exchanged Context Facet remains `DARK`.

Those audit records make experimental comparison possible. They do not make the exchanged Context Facet `LIT`. The exchanged Facet becomes `LIT` only when its resolving `cgvu:` URI is carried and available at the crossing.

Dark Uncertainty is the unrepresented decision exposure created when a workflow proceeds across operationally ignorant Context Facet crossings as though immediate sender intent and current reader interpretation were equivalent.

For the next finite experiment, the original source intent is fixed as `A`.

At each consecutive `DARK` crossing:

1. the current reader independently assigns either `A` or `B` with equal probability;
2. the reader’s interpretation is compared with the immediate sender’s intent; and
3. that interpretation becomes the reader’s intent if they send the artifact onward.

For a path of length $n$:

$$
P(\text{coherence maintained})
=
\left(\frac{1}{2}\right)^n
$$

$$
P(\text{at least one local flip})
=
1-\left(\frac{1}{2}\right)^n
$$

$$
P(\text{endpoint agreement})
=
\frac{1}{2}
$$

$$
P(
\text{coherence maintained}
\mid
\text{endpoint agreement}
)
=
2^{1-n}
$$

The following experiment exactly enumerates all $2^n$ reader-interpretation histories for every path length from 1 through 10.

These probabilities belong only to the independent, equally likely binary experiment. They are not organizational error-rate estimates or protocol properties.

````python
from fractions import Fraction
from itertools import product


SOURCE_REFERENT = "A"

BINARY_REFERENTS = (
    "A",
    "B",
)


def exact_probability_label(
    probability,
):
    """Format an exact probability as a fraction and percentage."""
    percent = (
        f"{float(probability) * 100:.6f}"
        .rstrip("0")
        .rstrip(".")
    )

    return (
        f"{probability.numerator}/"
        f"{probability.denominator} "
        f"({percent}%)"
    )


COMPOUNDING_TABLE = {
    "Dark crossings": [],
    "Possible histories": [],
    "Coherent histories": [],
    "P(coherence maintained)": [],
    "Histories with at least one local flip": [],
    "P(at least one local flip)": [],
    "Endpoint-agreeing histories": [],
    "P(endpoint agreement)": [],
    "Coherent among endpoint-agreeing histories": [],
}

COMPOUNDING_COUNTS = {}


for path_length in range(
    1,
    11,
):
    histories = tuple(
        product(
            BINARY_REFERENTS,
            repeat=path_length,
        )
    )

    coherent_histories = 0
    histories_with_flip = 0
    endpoint_agreeing_histories = 0
    coherent_and_endpoint_agreeing = 0

    for history in histories:
        immediate_sender_intent = (
            SOURCE_REFERENT
        )

        local_results = []

        for reader_interpretation in history:
            local_results.append(
                comparison_result(
                    immediate_sender_intent,
                    reader_interpretation,
                )
            )

            immediate_sender_intent = (
                reader_interpretation
            )

        coherent = all(
            result == MATCH
            for result in local_results
        )

        contains_flip = any(
            result == NO_MATCH
            for result in local_results
        )

        endpoint_agrees = (
            history[-1]
            == SOURCE_REFERENT
        )

        coherent_histories += coherent
        histories_with_flip += contains_flip
        endpoint_agreeing_histories += (
            endpoint_agrees
        )

        coherent_and_endpoint_agreeing += (
            coherent
            and endpoint_agrees
        )

    possible_histories = len(
        histories
    )

    p_coherent = Fraction(
        coherent_histories,
        possible_histories,
    )

    p_flip = Fraction(
        histories_with_flip,
        possible_histories,
    )

    p_endpoint = Fraction(
        endpoint_agreeing_histories,
        possible_histories,
    )

    p_coherent_given_endpoint = Fraction(
        coherent_and_endpoint_agreeing,
        endpoint_agreeing_histories,
    )

    assert (
        possible_histories
        == 2**path_length
    )

    assert coherent_histories == 1

    assert (
        p_coherent
        == Fraction(1, 2) ** path_length
    )

    assert (
        p_flip
        == 1
        - Fraction(1, 2) ** path_length
    )

    assert (
        p_endpoint
        == Fraction(1, 2)
    )

    assert (
        p_coherent_given_endpoint
        == Fraction(
            1,
            2 ** (path_length - 1),
        )
    )

    COMPOUNDING_COUNTS[
        path_length
    ] = {
        "possible": possible_histories,
        "coherent": coherent_histories,
        "flip": histories_with_flip,
        "endpoint": endpoint_agreeing_histories,
        "p_coherent": p_coherent,
        "p_flip": p_flip,
        "p_endpoint": p_endpoint,
        "p_coherent_given_endpoint": (
            p_coherent_given_endpoint
        ),
    }

    COMPOUNDING_TABLE[
        "Dark crossings"
    ].append(
        path_length
    )

    COMPOUNDING_TABLE[
        "Possible histories"
    ].append(
        possible_histories
    )

    COMPOUNDING_TABLE[
        "Coherent histories"
    ].append(
        coherent_histories
    )

    COMPOUNDING_TABLE[
        "P(coherence maintained)"
    ].append(
        exact_probability_label(
            p_coherent
        )
    )

    COMPOUNDING_TABLE[
        "Histories with at least one local flip"
    ].append(
        histories_with_flip
    )

    COMPOUNDING_TABLE[
        "P(at least one local flip)"
    ].append(
        exact_probability_label(
            p_flip
        )
    )

    COMPOUNDING_TABLE[
        "Endpoint-agreeing histories"
    ].append(
        endpoint_agreeing_histories
    )

    COMPOUNDING_TABLE[
        "P(endpoint agreement)"
    ].append(
        exact_probability_label(
            p_endpoint
        )
    )

    COMPOUNDING_TABLE[
        "Coherent among endpoint-agreeing histories"
    ].append(
        exact_probability_label(
            p_coherent_given_endpoint
        )
    )


ten_crossings = COMPOUNDING_COUNTS[
    10
]

assert (
    ten_crossings["p_coherent"]
    == Fraction(1, 1024)
)

assert (
    ten_crossings["p_flip"]
    == Fraction(1023, 1024)
)

assert (
    ten_crossings["p_endpoint"]
    == Fraction(1, 2)
)

assert (
    ten_crossings[
        "p_coherent_given_endpoint"
    ]
    == Fraction(1, 512)
)


print(
    "Complete enumeration for every binary "
    "path length from 1 through 10."
)

print_table(
    COMPOUNDING_TABLE
)

print(
    "At n=10: coherence maintained = 1/1,024; "
    "at least one local flip = 1,023/1,024; "
    "endpoint agreement = 1/2; "
    "coherence among endpoint-agreeing histories "
    "= 1/512."
)
````

### **Minimal Context-Blind Versus Context-Aware Proof**

This proof isolates only the `Meaning` Facet by holding the raw temperature Data and policy fixed.

The shared policy is:

```text
IF interpreted temperature is greater than 25°C
THEN GO
ELSE NO-GO
```

The Context-blind interface receives only the raw Data:

```python
context_blind_decide(28)
```

Because that interface does not receive a unit, it uses one explicitly documented implementation default:

```text
Celsius
```

The default is part of the interface definition. It is not presented as a universally correct interpretation.

The Context-aware interface receives the same raw Data plus a declared unit:

```python
context_aware_decide(28, unit)
```

It normalizes the supplied unit to Celsius and then calls the exact same policy function used by the Context-blind interface.

The complete declared comparison is:

```text
28 interpreted as Celsius    → 28°C    → GO
28 interpreted as Fahrenheit → -2.2°C  → NO-GO
```

This proof does not claim that a Context-blind engine is universally wrong. It establishes the narrower result:

> **When the same Data requires different verdicts under two declared interpretations, Data alone is insufficient to select the correct verdict across both cases.**

````python
# Minimal Context-blind versus Context-aware proof.

CONTEXT_BLIND_DEFAULT_UNIT = "Celsius"

DECLARED_TEMPERATURE_UNITS = (
    "Celsius",
    "Fahrenheit",
)

RAW_TEMPERATURE_DATA = 28


def normalize_to_celsius(
    recorded_temperature,
    unit,
):
    """Normalize a recorded temperature under its declared unit."""
    if unit == "Celsius":
        return float(recorded_temperature)

    if unit == "Fahrenheit":
        return (
            float(recorded_temperature) - 32
        ) * 5 / 9

    raise ValueError(
        f"undeclared temperature unit: {unit}"
    )


def shared_temperature_policy(
    interpreted_celsius,
):
    """Return GO only when interpreted temperature is above 25°C."""
    return (
        "GO"
        if interpreted_celsius > 25
        else "NO-GO"
    )


POLICY_FUNCTION = shared_temperature_policy


def context_blind_decide(
    recorded_temperature,
):
    """Apply the documented Celsius default and the shared policy."""
    interpreted_celsius = normalize_to_celsius(
        recorded_temperature,
        CONTEXT_BLIND_DEFAULT_UNIT,
    )

    return POLICY_FUNCTION(
        interpreted_celsius
    )


def context_aware_decide(
    recorded_temperature,
    unit,
):
    """Normalize the declared unit and apply the shared policy."""
    interpreted_celsius = normalize_to_celsius(
        recorded_temperature,
        unit,
    )

    return POLICY_FUNCTION(
        interpreted_celsius
    )


CONTEXT_AWARE_RESULTS = {
    unit: {
        "raw Data": RAW_TEMPERATURE_DATA,
        "interpreted Celsius": normalize_to_celsius(
            RAW_TEMPERATURE_DATA,
            unit,
        ),
        "Operational Verdict": context_aware_decide(
            RAW_TEMPERATURE_DATA,
            unit,
        ),
    }
    for unit in DECLARED_TEMPERATURE_UNITS
}

CONTEXT_BLIND_RESULT = context_blind_decide(
    RAW_TEMPERATURE_DATA
)


# The raw Data is identical under both declared interpretations.
assert {
    result["raw Data"]
    for result in CONTEXT_AWARE_RESULTS.values()
} == {
    RAW_TEMPERATURE_DATA
}


# Both interfaces call the same policy-function object.
assert "POLICY_FUNCTION" in (
    context_blind_decide.__code__.co_names
)

assert "POLICY_FUNCTION" in (
    context_aware_decide.__code__.co_names
)

assert (
    context_blind_decide.__globals__[
        "POLICY_FUNCTION"
    ]
    is context_aware_decide.__globals__[
        "POLICY_FUNCTION"
    ]
    is shared_temperature_policy
)


# The two declared interpretations construct different operational states.
assert (
    CONTEXT_AWARE_RESULTS["Celsius"][
        "interpreted Celsius"
    ]
    != CONTEXT_AWARE_RESULTS["Fahrenheit"][
        "interpreted Celsius"
    ]
)


# The two Context-aware Operational Verdicts differ.
assert {
    result["Operational Verdict"]
    for result in CONTEXT_AWARE_RESULTS.values()
} == {
    "GO",
    "NO-GO",
}


# The Context-blind interface is deterministic.
assert (
    context_blind_decide(
        RAW_TEMPERATURE_DATA
    )
    == CONTEXT_BLIND_RESULT
)


# One Context-blind output cannot equal both required outputs.
assert not all(
    CONTEXT_BLIND_RESULT
    == result["Operational Verdict"]
    for result in CONTEXT_AWARE_RESULTS.values()
)


print(
    "Complete declared Meaning state space: "
    "2 unit interpretations, not a sample."
)

for unit in DECLARED_TEMPERATURE_UNITS:
    result = CONTEXT_AWARE_RESULTS[unit]
    interpreted_celsius = result[
        "interpreted Celsius"
    ]

    temperature_label = (
        f"{interpreted_celsius:.0f}°C"
        if interpreted_celsius.is_integer()
        else f"{interpreted_celsius:.1f}°C"
    )

    print(
        f"{RAW_TEMPERATURE_DATA} interpreted as "
        f"{unit:<10} → "
        f"{temperature_label:<7} → "
        f"{result['Operational Verdict']}"
    )

print(
    "Context-blind default "
    f"({CONTEXT_BLIND_DEFAULT_UNIT}): "
    f"{CONTEXT_BLIND_RESULT}"
)
````

### **Interpretation Is More Than Semantics**

The minimal temperature proof isolates only the `Meaning` Facet.

The full Context gauge is:

$$
C=(M,S,W)
$$

where:

- \(M\) is `Meaning`;
- \(S\) is `Structure`; and
- \(W\) is `World`.

An interpretation is the referent assigned by the current reader. The operational state constructed from Data can therefore be expressed as:

$$
\text{Operational State}
=
\phi(\text{Data},M,S,W)
$$

Identical Data can produce different operational states because of differences in any of the three Context dimensions:

| Context Facet | Identical or selected Data | Declared interpretation difference | Operational effect |
| --- | --- | --- | --- |
| `Meaning` | recorded temperature `28` | Celsius or Fahrenheit | changes the normalized temperature |
| `Structure` | recorded temperature `29` | rounded up or rounded down | changes the coldest plausible temperature |
| `World` | the selected temperature record | Baltic Sea or Atlantic Ocean | changes the declared freezing threshold |
| `Structure` | the dates in `table-1` | `MM/DD/YYYY` or `DD/MM/YYYY` | changes which record represents May |

Two readers can agree on `Meaning` while applying different `Structure` or `World` interpretations:

$$
D_A=D_B
$$

and:

$$
M_A=M_B
$$

while:

$$
(S_A,W_A)\ne(S_B,W_B)
$$

The resulting operational states can therefore differ:

$$
\phi(D_A,M_A,S_A,W_A)
\ne
\phi(D_B,M_B,S_B,W_B)
$$

If the shared policy maps those states to different outputs, the Operational Verdict changes even though the policy itself remains internally consistent.

> **Semantic agreement does not establish complete Context agreement. Differences in Meaning, Structure, or World can change the operational state supplied to a Decision Engine.**

### **Maritime Decision Engine: Complete Rule Documentation**

The maritime experiment uses one deterministic Decision Engine whose rules are fully declared below.

The Query rule first selects the record associated with May. The remaining rules then transform that record into an interpreted operational state and return the captain’s Operational Verdict.

The Decision Engine does not measure `DARK` or `LIT`, compare intent with interpretation, or produce a Governor Verdict. It receives declared interpretation inputs and applies the same deterministic rules to them.

#### Declared Experimental Constants

This finite experiment declares:

- a Baltic Sea freezing threshold of `31.0°F`;
- an Atlantic Ocean freezing threshold of `28.6°F`; and
- a Fahrenheit rounding band of `0.5°F`.

These values are controlled constants of this experiment. They are not presented as universal physical or navigational facts.

#### Complete Rule Mapping

| Protocol input | Python variable | Function | Effect |
| --- | --- | --- | --- |
| `date.structure` | `is_mm_dd` | `month_of`, `select_row` | selects the record |
| `temperature.meaning` | `is_celsius` | `to_fahrenheit` | normalizes the value |
| `temperature.structure` | `is_rounded_up` | `floor_of` | calculates the lower bound |
| `temperature.world` | `is_baltic` | `freezing_line` | selects the threshold |
| interpreted operational state | function arguments | `decide` | returns `GO` or `NO-GO` |

#### Query Rule

`date.structure` determines which record represents May.

```text
Is date.structure MM/DD/YYYY?
├── Yes: month_of uses the first date number
│         └── select the record whose first number is 5
└── No:  month_of uses the second date number
          └── select the record whose second number is 5
```

#### Meaning Rule

`temperature.meaning` determines whether the recorded value is already Fahrenheit or must be converted from Celsius.

```text
Is temperature.meaning Celsius?
├── Yes: convert the recorded value to Fahrenheit
└── No:  interpret the recorded value as Fahrenheit
```

#### Structure Rule

`temperature.structure` determines the coldest temperature consistent with the recorded value under the experiment’s declared rounding band.

```text
Was the recorded temperature rounded up?
├── Yes: coldest plausible = interpreted Fahrenheit − 0.5°F
└── No:  coldest plausible = interpreted Fahrenheit
```

#### World Rule

`temperature.world` selects the applicable declared freezing threshold.

```text
Does temperature.world identify the Baltic Sea?
├── Yes: selected threshold = 31.0°F
└── No:  selected threshold = 28.6°F
```

#### Verdict Rule

The Operational Verdict is `GO` only when the coldest plausible temperature is strictly above the selected freezing threshold.

```text
Is coldest plausible temperature > selected freezing threshold?
├── Yes: GO
└── No:  NO-GO
          └── equality therefore produces NO-GO
```

````python
"""Deterministic maritime Decision Engine.

The engine receives a recorded temperature and three declared
temperature-interpretation inputs. It constructs the interpreted
operational state and returns an Operational Verdict.
"""


class MaritimeExperimentConstants:
    """Constants declared only for this finite maritime experiment."""

    FREEZE_BALTIC_F = 31.0
    FREEZE_ATLANTIC_F = 28.6
    ROUNDING_BAND_F = 0.5


def to_fahrenheit(
    recorded_temperature,
    is_celsius,
):
    """Apply the declared temperature Meaning interpretation."""
    return (
        recorded_temperature * 9 / 5 + 32
        if is_celsius
        else float(recorded_temperature)
    )


def floor_of(
    interpreted_fahrenheit,
    is_rounded_up,
):
    """Apply the declared Structure interpretation to find the lower bound."""
    return (
        interpreted_fahrenheit
        - MaritimeExperimentConstants.ROUNDING_BAND_F
        if is_rounded_up
        else interpreted_fahrenheit
    )


def freezing_line(
    is_baltic,
):
    """Select the threshold declared by the World interpretation."""
    return (
        MaritimeExperimentConstants.FREEZE_BALTIC_F
        if is_baltic
        else MaritimeExperimentConstants.FREEZE_ATLANTIC_F
    )


def decide(
    recorded_temperature,
    is_celsius,
    is_rounded_up,
    is_baltic,
):
    """Return GO only when the coldest plausible value exceeds the threshold."""
    interpreted_fahrenheit = to_fahrenheit(
        recorded_temperature,
        is_celsius,
    )

    coldest_plausible = floor_of(
        interpreted_fahrenheit,
        is_rounded_up,
    )

    freezing_threshold = freezing_line(
        is_baltic,
    )

    return (
        "GO"
        if coldest_plausible > freezing_threshold
        else "NO-GO"
    )


def trace_decision(
    recorded_temperature,
    is_celsius,
    is_rounded_up,
    is_baltic,
):
    """Expose every operational value used to produce the Verdict."""
    interpreted_fahrenheit = to_fahrenheit(
        recorded_temperature,
        is_celsius,
    )

    coldest_plausible = floor_of(
        interpreted_fahrenheit,
        is_rounded_up,
    )

    freezing_threshold = freezing_line(
        is_baltic,
    )

    operational_verdict = decide(
        recorded_temperature,
        is_celsius,
        is_rounded_up,
        is_baltic,
    )

    return {
        "recorded temperature": recorded_temperature,
        "interpreted Fahrenheit temperature": interpreted_fahrenheit,
        "coldest plausible Fahrenheit temperature": coldest_plausible,
        "selected freezing threshold": freezing_threshold,
        "Operational Verdict": operational_verdict,
    }


MARITIME_TRACE_FIELDS = {
    "recorded temperature",
    "interpreted Fahrenheit temperature",
    "coldest plausible Fahrenheit temperature",
    "selected freezing threshold",
    "Operational Verdict",
}


# This declared boundary case produces equality:
# 31.5°F rounded up gives a 31.0°F lower bound,
# equal to the declared Baltic threshold of 31.0°F.
MARITIME_BOUNDARY_TRACE = trace_decision(
    recorded_temperature=31.5,
    is_celsius=False,
    is_rounded_up=True,
    is_baltic=True,
)


# The engine is deterministic.
assert MARITIME_BOUNDARY_TRACE == trace_decision(
    recorded_temperature=31.5,
    is_celsius=False,
    is_rounded_up=True,
    is_baltic=True,
)


# The trace exposes every required operational value.
assert (
    set(MARITIME_BOUNDARY_TRACE)
    == MARITIME_TRACE_FIELDS
)


# Equality produces NO-GO.
assert (
    MARITIME_BOUNDARY_TRACE[
        "coldest plausible Fahrenheit temperature"
    ]
    == MARITIME_BOUNDARY_TRACE[
        "selected freezing threshold"
    ]
    == 31.0
)

assert (
    MARITIME_BOUNDARY_TRACE[
        "Operational Verdict"
    ]
    == "NO-GO"
)


# A value strictly above the threshold produces GO.
assert decide(
    recorded_temperature=31.6,
    is_celsius=False,
    is_rounded_up=True,
    is_baltic=True,
) == "GO"


print(
    "Maritime Decision Engine verified: deterministic, "
    "fully traced, and equality returns NO-GO."
)
````

````python
# Exact Query-rule proof for query-1.
import re


MONTHS = {
    "january": 1,
    "february": 2,
    "march": 3,
    "april": 4,
    "may": 5,
    "june": 6,
    "july": 7,
    "august": 8,
    "september": 9,
    "october": 10,
    "november": 11,
    "december": 12,
}

DECLARED_DATE_STRUCTURES = (
    ("MM/DD/YYYY", True),
    ("DD/MM/YYYY", False),
)


def month_of(
    date_string,
    is_mm_dd,
):
    """Return the interpreted month under one declared date structure."""
    first, second, _year = (
        int(part)
        for part in date_string.split("/")
    )

    return first if is_mm_dd else second


def select_row(
    table,
    month_name,
    is_mm_dd,
):
    """Return the first record matching the interpreted month."""
    requested_month = MONTHS[
        month_name.lower()
    ]

    for row_index, date_string in enumerate(
        table["date"]
    ):
        if (
            month_of(
                date_string,
                is_mm_dd,
            )
            == requested_month
        ):
            return row_index

    return None


def month_named(
    query_text,
):
    """Return the one month explicitly named by the Query."""
    named_months = [
        word
        for word in re.findall(
            r"[a-z]+",
            query_text.lower(),
        )
        if word in MONTHS
    ]

    assert len(named_months) == 1, (
        "query-1 must name exactly one month; "
        f"found {named_months}"
    )

    return named_months[0].capitalize()


QUERY_MONTH = month_named(
    query_1["text"][0]
)

QUERY_SELECTIONS = {}

for (
    date_structure,
    is_mm_dd,
) in DECLARED_DATE_STRUCTURES:
    selected_row = select_row(
        db.get(1),
        QUERY_MONTH,
        is_mm_dd,
    )

    assert selected_row is not None

    QUERY_SELECTIONS[
        date_structure
    ] = {
        "selected row": selected_row + 1,
        "selected date": db.get(1)[
            "date"
        ][selected_row],
        "selected location": db.get(1)[
            "location"
        ][selected_row],
        "recorded temperature": db.get(1)[
            "temperature"
        ][selected_row],
    }


# query-1 names exactly one month.
assert QUERY_MONTH == "May"


# Every declared date structure was evaluated exactly once.
assert tuple(
    QUERY_SELECTIONS
) == tuple(
    date_structure
    for (
        date_structure,
        _is_mm_dd,
    ) in DECLARED_DATE_STRUCTURES
)

assert len(QUERY_SELECTIONS) == 2


# Both interpretations select the expected records.
assert QUERY_SELECTIONS[
    "MM/DD/YYYY"
] == {
    "selected row": 1,
    "selected date": "05/04/2026",
    "selected location": "Paris",
    "recorded temperature": 29,
}

assert QUERY_SELECTIONS[
    "DD/MM/YYYY"
] == {
    "selected row": 2,
    "selected date": "04/05/2026",
    "selected location": "Cambridge",
    "recorded temperature": 28,
}


# The two declared interpretations select distinct records.
assert (
    QUERY_SELECTIONS[
        "MM/DD/YYYY"
    ]["selected row"]
    != QUERY_SELECTIONS[
        "DD/MM/YYYY"
    ]["selected row"]
)


print(
    "Complete declared date-structure state space: "
    "2 interpretations, not a sample."
)

print_table({
    "date.structure": list(
        QUERY_SELECTIONS
    ),
    "selected row": [
        result["selected row"]
        for result in QUERY_SELECTIONS.values()
    ],
    "selected date": [
        result["selected date"]
        for result in QUERY_SELECTIONS.values()
    ],
    "selected location": [
        result["selected location"]
        for result in QUERY_SELECTIONS.values()
    ],
    "recorded temperature": [
        result["recorded temperature"]
        for result in QUERY_SELECTIONS.values()
    ],
})

print(
    f'Query: "What was the temperature and '
    f'location in {QUERY_MONTH}?"'
)
````

### **Regret Under a Fixed Finite Scenario Set**

Regret is measured against the experiment’s recorded source-intent state, which serves as the fixed baseline for each scenario.

The recorded source intent is not presented as metaphysical or external truth. It is the interpretation state recorded by the experiment when the artifact-field Context Facets are authored.

For this proof:

- **recorded source-intent state** is one of the experiment’s 16 complete interpretation states;
- **source-intent Operational Verdict** is the maritime Decision Engine output under that recorded baseline;
- **reader-interpreted state** is one of the same 16 declared interpretation states assigned by the reader;
- **reader-interpreted Operational Verdict** is the Decision Engine output under the reader’s state;
- **realized cash** is the payoff produced by the reader-interpreted Operational Verdict under the recorded source-intent baseline;
- **optimal cash** is the best payoff under that recorded source-intent baseline; and
- **regret** is:

$$
\text{regret}
=
\text{optimal cash}
-
\text{realized cash}
$$

The fixed payoff ledger is:
| Reader Operational Verdict | Source-intent operational condition | Realized cash | Optimal cash | Regret |
| --- | --- | ---: | ---: | ---: |
| `GO` | source-intent Verdict is `GO` | +\$1,000,000 | +\$1,000,000 | \$0 |
| `NO-GO` | source-intent Verdict is `NO-GO` | \$0 | \$0 | \$0 |
| `NO-GO` | source-intent Verdict is `GO` | \$0 | +\$1,000,000 | \$1,000,000 |
| `GO` | source-intent Verdict is `NO-GO` | −\$2,000,000 | \$0 | \$2,000,000 |

The Python cell exists to measure the exact decision and economic consequences of pairing every recorded source-intent state with every reader-interpreted state.

It receives the 16 states and deterministic results established by the complete Track 1-C state proof.

It calculates the Cartesian product:

$$
16 \times 16 = 256
$$

and reports:

- number of recorded source-intent states;
- number of reader-interpreted states;
- total fixed scenarios;
- Operational Verdict mismatches;
- total realized cash;
- total optimal cash; and
- total regret.

The proof is exhaustive over one fixed finite scenario set. It is not sampled and does not compare different random fleets.

Its assertions verify that:

- exactly 256 unique state pairs exist;
- all 16 recorded source-intent states appear;
- all 16 reader-interpreted states appear;
- regret is never negative; and
- regret is zero whenever the two Operational Verdicts match.

````python
# Exact Track 1-C regret proof over one fixed 16 x 16 scenario set.

TRACK1C_PAYOFF_LEDGER = {
    ("GO", "GO"): (
        1_000_000,
        1_000_000,
        0,
    ),
    ("NO-GO", "NO-GO"): (
        0,
        0,
        0,
    ),
    ("NO-GO", "GO"): (
        0,
        1_000_000,
        1_000_000,
    ),
    ("GO", "NO-GO"): (
        -2_000_000,
        0,
        2_000_000,
    ),
}


def regret_outcome(
    reader_operational_verdict,
    source_intent_operational_verdict,
):
    """Return realized cash, optimal cash, and regret."""
    return TRACK1C_PAYOFF_LEDGER[
        (
            reader_operational_verdict,
            source_intent_operational_verdict,
        )
    ]


TRACK1C_REGRET_SCENARIOS = []

for scenario_number, (
    source_intent_state,
    reader_interpreted_state,
) in enumerate(
    product(
        TRACK1C_INTERPRETATION_STATES,
        repeat=2,
    ),
    1,
):
    source_intent_result = TRACK1C_STATE_RESULTS[
        source_intent_state
    ]

    reader_interpreted_result = TRACK1C_STATE_RESULTS[
        reader_interpreted_state
    ]

    source_intent_operational_verdict = (
        source_intent_result[
            "Operational Verdict"
        ]
    )

    reader_interpreted_operational_verdict = (
        reader_interpreted_result[
            "Operational Verdict"
        ]
    )

    (
        realized_cash,
        optimal_cash,
        regret,
    ) = regret_outcome(
        reader_interpreted_operational_verdict,
        source_intent_operational_verdict,
    )

    TRACK1C_REGRET_SCENARIOS.append({
        "scenario identifier": f"R{scenario_number:03d}",
        "recorded source-intent state": (
            source_intent_result[
                "state identifier"
            ]
        ),
        "reader-interpreted state": (
            reader_interpreted_result[
                "state identifier"
            ]
        ),
        "source-intent Operational Verdict": (
            source_intent_operational_verdict
        ),
        "reader-interpreted Operational Verdict": (
            reader_interpreted_operational_verdict
        ),
        "realized cash": realized_cash,
        "optimal cash": optimal_cash,
        "regret": regret,
    })


TRACK1C_STATE_PAIRS = {
    (
        scenario["recorded source-intent state"],
        scenario["reader-interpreted state"],
    )
    for scenario in TRACK1C_REGRET_SCENARIOS
}

TRACK1C_SOURCE_INTENT_STATE_IDS = {
    scenario["recorded source-intent state"]
    for scenario in TRACK1C_REGRET_SCENARIOS
}

TRACK1C_READER_INTERPRETED_STATE_IDS = {
    scenario["reader-interpreted state"]
    for scenario in TRACK1C_REGRET_SCENARIOS
}


assert len(TRACK1C_REGRET_SCENARIOS) == 256
assert len(TRACK1C_STATE_PAIRS) == 256
assert len(TRACK1C_SOURCE_INTENT_STATE_IDS) == 16
assert len(TRACK1C_READER_INTERPRETED_STATE_IDS) == 16

assert all(
    scenario["regret"] >= 0
    for scenario in TRACK1C_REGRET_SCENARIOS
)

assert all(
    scenario["regret"] == 0
    for scenario in TRACK1C_REGRET_SCENARIOS
    if (
        scenario["source-intent Operational Verdict"]
        == scenario["reader-interpreted Operational Verdict"]
    )
)


TRACK1C_VERDICT_MISMATCHES = sum(
    scenario["source-intent Operational Verdict"]
    != scenario["reader-interpreted Operational Verdict"]
    for scenario in TRACK1C_REGRET_SCENARIOS
)

TRACK1C_TOTAL_REALIZED_CASH = sum(
    scenario["realized cash"]
    for scenario in TRACK1C_REGRET_SCENARIOS
)

TRACK1C_TOTAL_OPTIMAL_CASH = sum(
    scenario["optimal cash"]
    for scenario in TRACK1C_REGRET_SCENARIOS
)

TRACK1C_TOTAL_REGRET = sum(
    scenario["regret"]
    for scenario in TRACK1C_REGRET_SCENARIOS
)

assert (
    TRACK1C_TOTAL_REGRET
    == TRACK1C_TOTAL_OPTIMAL_CASH
    - TRACK1C_TOTAL_REALIZED_CASH
)


TRACK1C_REGRET_SCENARIO_TABLE = {
    column: [
        (
            f"${scenario[column]:,}"
            if column in {
                "realized cash",
                "optimal cash",
                "regret",
            }
            else scenario[column]
        )
        for scenario in TRACK1C_REGRET_SCENARIOS
    ]
    for column in (
        "scenario identifier",
        "recorded source-intent state",
        "reader-interpreted state",
        "source-intent Operational Verdict",
        "reader-interpreted Operational Verdict",
        "realized cash",
        "optimal cash",
        "regret",
    )
}


print(
    "Complete fixed scenario set: "
    "16 recorded source-intent states × "
    "16 reader-interpreted states = "
    "256 state pairs, not a sample."
)

print_table(TRACK1C_REGRET_SCENARIO_TABLE)

print_table({
    "Exact regret measure": [
        "Recorded source-intent states",
        "Reader-interpreted states",
        "Total fixed scenarios",
        "Operational Verdict mismatches",
        "Total realized cash",
        "Total optimal cash",
        "Total regret",
    ],
    "Result": [
        len(TRACK1C_SOURCE_INTENT_STATE_IDS),
        len(TRACK1C_READER_INTERPRETED_STATE_IDS),
        len(TRACK1C_REGRET_SCENARIOS),
        TRACK1C_VERDICT_MISMATCHES,
        f"${TRACK1C_TOTAL_REALIZED_CASH:,}",
        f"${TRACK1C_TOTAL_OPTIMAL_CASH:,}",
        f"${TRACK1C_TOTAL_REGRET:,}",
    ],
})
````

## Track 1-D: Resolving Dark Context

Track 1-D tests the protocol’s explicit routing and resolution cycle at one declared Observatron boundary.

The experiment keeps four namespaces separate:

| Namespace | Values | Purpose |
| --- | --- | --- |
| Context state | `DARK`, `LIT` | whether a resolving `cgvu:` URI is carried |
| comparison result | `MATCH`, `NO MATCH` | whether reader interpretation and immediate sender intent point to the same referent |
| Governor Verdict | `ACT`, `ASK`, `HALT` | which independently declared Intent Map instruction is selected |
| Operational Verdict | `GO`, `NO-GO` | the captain’s maritime decision |

A `DARK` Context Facet carries no resolving `cgvu:` URI at the observed crossing. A `LIT` Context Facet carries a resolving `cgvu:` URI.

`LIT` does not automatically mean `MATCH`. In this experiment, selected inputs are deliberately bound to their recorded source-intent referents so that the resulting downstream effect can be measured against the Track 1-C baseline.

The Governor is a decision model for routing. It evaluates an event against explicit Intent Map rules and returns:

- a Governor Verdict: `ACT`, `ASK`, or `HALT`; and
- a Decision Route identifying the registered Handler.

The Governor does not choose an interpretation, bind a referent, or infer that `LIT` means `ACT`.

The distinction between `ASK` and `ACT` is essential. They are independent Governor instructions with no required lifecycle connection.

When the Governor returns `ASK`, it dispatches the registered `ASK` Handler. That Handler owns the complete interaction: requesting an answer, receiving the response, and performing whatever response behavior is declared for that Handler.

In this experiment, the `ASK` Handler uses the response to bind the resolving `cgvu:` URI for `temperature.meaning`. This changes the Context Facet from `DARK` to `LIT`. No `ACT` Verdict is produced or required at any point in that process.

`ACT` remains an entirely independent instruction. For example, a separate Intent Map rule could observe the word `foo`, return `ACT`, and route to a Handler that displays the notification `bar`. That example has no connection to the `ASK` interaction or the Context-state transition demonstrated here.

For the single-facet routing proof, the experiment records the source intent for `temperature.meaning` as the declared Celsius referent. This is an experimental baseline, not a claim of external or metaphysical truth.

### **Single-Facet `ASK` Routing and Resolution Proof**

This proof demonstrates how one unresolved Context Facet is routed to an addressable Handler and resolved at one declared Observatron boundary.

It focuses on the `Meaning` facet of the `temperature` Spike from `table-1`. The proof begins with that facet explicitly reset to `DARK`, meaning it carries no resolving `cgvu:` URI.

The local Track 1-D Intent Map rule matches an event with all four of these properties:

- event type: `context-facet-observed`;
- Data: `temperature`;
- Context Facet: `meaning`; and
- Context state: `DARK`.

When the event matches, the Governor returns:

- Governor Verdict: `ASK`; and
- Decision Route: `cgp:/track-1-d/handlers/ask-temperature-meaning`.

The Decision Route is resolved through the Handler Registry. The registered `ASK` Handler owns the rest of the interaction: it presents the declared Celsius/Fahrenheit answer space, receives the selected response, validates that the selected `cgvu:` URI belongs to that answer space, and binds it to the Context Facet.

After confirmation, the cell displays:

This matters because it keeps the protocol’s responsibilities explicit:

The complete interaction belongs to the `ASK` Handler. It does not produce, require, or imply an `ACT` Verdict.

This is an exact single-facet routing proof, not a sampled decision-impact experiment. The later fixed-scenario comparison measures how resolved inputs affect Operational Verdicts and regret.

````python
# Exact single-facet ASK routing and resolution proof.
import ipywidgets as widgets


TRACK1D_FACET = "meaning"
TRACK1D_ASK_ROUTE = (
    "cgp:/track-1-d/handlers/"
    "ask-temperature-meaning"
)

TRACK1D_SPIKE = next(
    spike["spike"]
    for spike in observatron_1.spikes
    if (
        spike["artifact"] == "table-1"
        and spike["data"] == "temperature"
    )
)

TRACK1D_INTENT_RULE = {
    "when": {
        "event type": "context-facet-observed",
        "data": "temperature",
        "facet": TRACK1D_FACET,
        "Context state": "DARK",
    },
    "then": {
        "Governor Verdict": "ASK",
        "Decision Route": TRACK1D_ASK_ROUTE,
    },
}


def track1d_governor(obs, spike, facet):
    """Return the instruction declared for the observed event."""
    spike_record = next(
        record
        for record in obs.spikes
        if record["spike"] == spike
    )

    event = {
        "event type": "context-facet-observed",
        "data": spike_record["data"],
        "facet": facet,
        "Context state": obs.state(spike, facet),
    }

    rule_matches = all(
        event[key] == expected
        for key, expected
        in TRACK1D_INTENT_RULE["when"].items()
    )

    if not rule_matches:
        return None

    return dict(TRACK1D_INTENT_RULE["then"])


def ask_temperature_meaning_handler(obs, spike, facet):
    """Own the complete ASK interaction and its declared resolution."""
    data = next(
        record["data"]
        for record in obs.spikes
        if record["spike"] == spike
    )

    labels = LOOP_CHOICES[data][facet]
    addresses = candidate_uris(
        LOOP_CHOICES,
        data,
        facet,
    )

    question = widgets.Label(
        "Is `temperature` Celsius or Fahrenheit?"
    )

    choice = widgets.RadioButtons(
        options=list(zip(labels, addresses)),
        layout=widgets.Layout(
            width="max-content"
        ),
    )

    confirm = widgets.Button(
        description="Confirm",
        button_style="primary",
    )

    result = widgets.Output()

    def on_confirm(_):
        result.clear_output()

        with result:
            selected_uri = choice.value

            assert selected_uri in addresses

            state_before = obs.state(
                spike,
                facet,
            )

            obs.facets[(spike, facet)] = (
                selected_uri
            )

            state_after = obs.state(
                spike,
                facet,
            )

            assert state_before == "DARK"
            assert state_after == "LIT"

            print(
                "ASK Handler received response: "
                f"{selected_uri}"
            )
            print(
                "Context-state transition: "
                f"{state_before} -> {state_after}"
            )

            print_table(obs.table())

            counts = obs.counts()

            print(
                f"{counts['lit']} LIT, "
                f"{counts['dark']} DARK of "
                f"{counts['context_facets']} "
                "Context Facets"
            )

        confirm.disabled = True
        choice.disabled = True

    confirm.on_click(on_confirm)

    return widgets.VBox([
        question,
        choice,
        confirm,
        result,
    ])


TRACK1D_HANDLER_REGISTRY = {
    TRACK1D_ASK_ROUTE: (
        ask_temperature_meaning_handler
    ),
}


# Reset the selected facet so the proof starts DARK when rerun.
observatron_1.facets[
    (
        TRACK1D_SPIKE,
        TRACK1D_FACET,
    )
] = ""


track1d_instruction = track1d_governor(
    observatron_1,
    TRACK1D_SPIKE,
    TRACK1D_FACET,
)

assert track1d_instruction == {
    "Governor Verdict": "ASK",
    "Decision Route": TRACK1D_ASK_ROUTE,
}

print_table({
    "Governor Verdict": [
        track1d_instruction[
            "Governor Verdict"
        ]
    ],
    "Decision Route": [
        track1d_instruction[
            "Decision Route"
        ]
    ],
})

track1d_handler = TRACK1D_HANDLER_REGISTRY[
    track1d_instruction["Decision Route"]
]

display(
    track1d_handler(
        observatron_1,
        TRACK1D_SPIKE,
        TRACK1D_FACET,
    )
)
````

### **Track 1-D Summary: Routing, Resolution, and Measured Outcome**

Track 1-D separates four protocol responsibilities:

| Component | Responsibility |
| --- | --- |
| Observatron | measures the event and the current Context state |
| Governor | matches the event against the Intent Map and returns a Governor Verdict plus a Decision Route |
| Handler Registry | resolves the Decision Route to its registered Handler |
| Handler | performs the complete behavior registered to that route |

The three Governor Verdicts are independent instructions:

- `ASK` routes to a Handler whose declared behavior involves requesting something;
- `ACT` routes to a Handler whose declared behavior performs an action; and
- `HALT` routes to a Handler whose declared behavior stops the applicable process.

There is no required sequence from one Governor Verdict to another. In particular, an `ASK` interaction does not produce or imply `ACT`.

The behavior belongs to the Handler. An `ASK` Handler may present a question, receive the response, and perform whatever response behavior is registered to that route. In the single-facet proof, the registered Handler validates the selected `cgvu:` URI and binds it to `temperature.meaning`, changing that Context Facet from `DARK` to `LIT`.

That binding behavior is specific to the demonstrated `ASK` route. It is not part of the definition of `ASK`.

Likewise, `ACT` does not inherently mean “bind a referent.” For example, an Intent Map rule could observe the word `foo`, return `ACT`, and route to a Handler that displays the notification `bar`. That independent action requires no preceding `ASK` and no Context-state transition.

The address namespaces also remain separate:

- a `cgp:` Decision Route identifies registered Handler behavior; and
- a `cgvu:` URI identifies the referent carried by a `LIT` Context Facet.

The cumulative outcome proof does not simulate additional Governor interactions. It re-evaluates the same 256 fixed Track 1-C scenario pairs after progressively binding four selected Context Facets to their recorded source-intent referents.

Its exact results are:

These results establish two separate Track 1-D claims:

1. a declared `ASK` route can resolve a `DARK` Context Facet entirely through its registered Handler; and

# Track 2 — Decision Shapes

## What Decisions Depend on, and What it Costs to Find Out

Not every `Context Facet` is consequential to every `Operational Decision`.

A **`Decision Shape`** is the set of `Essential Facet`s for a particular `Operational Decision`. It reveals which facets can change that decision's `Verdict`, and which facets are `Inert` with respect to that decision.

```text
All Context Facets for an Operational Decision
                    │
          ┌─────────┴─────────┐
          ▼                   ▼
      Essential             Inert
          │                   │
          ▼                   ▼
   Decision Shape      outside Decision Shape
```

An **`Essential Facet`** is a `Context Facet` whose interpretation can change the `Verdict` of that `Operational Decision`.

An **`Inert Facet`** is a `Context Facet` whose interpretation cannot change the `Verdict` of that `Operational Decision`.

This distinction matters because a `Context Facet` can change how data is interpreted without being consequential to the particular decision being made. Decision Shape analysis separates changes in interpretation from changes that can actually alter the decision's `Verdict`.

Decision Shapes do not make the decision, resolve the decision, or govern routing. They reveal the shape of the essential facets for an `Operational Decision` and identify the facets that are inert relative to it.

## Declared Candidate Facets for This Operational Decision

Before we can discover a `Decision Shape`, we need to declare where ambiguity is actually known to exist.

The three data fields in this example — `temperature`, `location`, and `date` — each have `Meaning`, `Structure`, and `World` facets. This gives us **9 `Context Facet`s in the frame**.

However, the existence of a facet does not mean that the facet is ambiguous.

For this `Operational Decision`, the `Intent Map` identifies the facets for which more than one interpretation is currently plausible. These become **candidate facets** for Decision Shape analysis.

A facet with no alternate candidate still exists, but it is not varied in this experiment.

## The Known Decision Rule

The `Decision Shape` experiment does not invent a new decision.

The captain's `Operational Decision` and its `Decision Rule` are already known.

```text
GO
or
NO-GO
```

The decision is evaluated from the interpreted data in four steps.

The raw date values are strings such as:

```text

```

The interpretation of `date.structure` determines how that value is parsed:

```text
MM/DD/YYYY
or
DD/MM/YYYY
```

That interpretation determines which row of the table is treated as the May row used by the decision.

### 2. Interpret the Temperature Value

The selected row contains a numeric temperature.

`temperature.meaning` determines whether that number is interpreted as:

```text
Celsius
or
Fahrenheit
```

The interpreted value is converted into Fahrenheit for comparison.

We call this value:

```text
temp_f
```

### 3. Determine the Coldest Temperature Consistent With the Record

`temperature.structure` determines how the recorded value was rounded:

```text
rounded up
or
rounded down
```

That interpretation determines the coldest plausible temperature consistent with the recorded value.

We call this:

```text
floor
```

### 4. Determine the Applicable Freezing Threshold

`temperature.world` determines which environmental freezing condition applies:

```text
Baltic Sea
or
Atlantic Ocean
```

That determines the applicable freezing threshold.

We call this:

```text
freezing_line
```

The final `Decision Rule` is:

```text
if floor > freezing_line:
    Verdict = GO
else:
    Verdict = NO-GO
```

The complete decision path is therefore:

```text
Raw table
   │
   ▼

   │
   ├── date.structure
   │       │
   │       ▼

   │

   │       │
   │       ▼

   │
   ├── temperature.structure
   │       │
   │       ▼
   │      floor
   │
   └── temperature.world
           │
           ▼
      freezing_line
           │
           ▼
   floor > freezing_line?
        /          \
      yes          no
       │            │
       ▼            ▼
      GO          NO-GO
```

```text
05/04/2026
  ├── date the reading was collected
  └── date the reported lowest temperature occurred
```

These interpretations change the contextual meaning of the artifact even if they do not necessarily change the fields used by the captain's `Decision Rule`.

That distinction is exactly what the Decision Shape test is designed to measure.

The experiment must therefore follow this order:

```text
6 candidate Context Facets
            │
            ▼
apply one complete interpretation state
            │
            ▼
materialize the interpreted artifact
            │
            ▼
apply the known Decision Rule
            │
            ▼
          Verdict
```

Only after all six candidate facets have participated in interpretation do we ask:

> If one facet changes while the other five remain fixed, can the `Verdict` change?

If yes, that facet is `Essential`.

If the interpretation changes but the `Verdict` never changes, that facet is `Inert` for this particular `Operational Decision`.

### Decision Rule as Code

The same `Decision Rule` can be expressed directly as executable logic.

```python
def decision_rule(
    raw_temperature,
    is_celsius,
    is_rounded_up,
    is_baltic
):
    # Meaning: interpret the recorded number.
    if is_celsius:
        temp_f = raw_temperature * 9 / 5 + 32
    else:
        temp_f = raw_temperature

    # Structure: determine the coldest plausible temperature.
    if is_rounded_up:
        floor = temp_f - 0.5
    else:
        floor = temp_f

    # World: determine the applicable freezing threshold.
    if is_baltic:
        freezing_line = 31.0
    else:
        freezing_line = 28.6

    # Operational Decision.
    if floor > freezing_line:
        return "GO"

    return "NO-GO"
```

So the final decision itself reduces to:

```python
if floor > freezing_line:
    Verdict = "GO"
else:
    Verdict = "NO-GO"
```

The variables leading into that comparison come from different `Context Facet`s:

```text
temperature.meaning
        │
        ▼
      temp_f
        │
temperature.structure
        │
        ▼
      floor ───────────────┐
                           │
temperature.world          │
        │                  │
        ▼                  │
 freezing_line ────────────┘
        │
        ▼
floor > freezing_line?
     /          \
   True        False
    │             │
    ▼             ▼
   GO           NO-GO
```

`date.structure` acts one step earlier because it determines **which row supplies `raw_temperature`**:

```python
if date_structure == "MM/DD/YYYY":
    may_row = row_where_first_date_number_is_5
else:
    may_row = row_where_second_date_number_is_5

raw_temperature = may_row["temperature"]

Verdict = decision_rule(
    raw_temperature,
    is_celsius,
    is_rounded_up,
    is_baltic
)
```

This distinction is important for the Decision Shape experiment:

- `date.structure` can change **which temperature enters the decision**.
- `temperature.meaning` can change the interpreted temperature.
- `temperature.structure` can change the plausible lower bound.
- `temperature.world` can change the freezing threshold.
- `location.world` and `date.meaning` still change the interpreted context, but the experiment will determine whether those changes can propagate far enough to change the `Verdict`.

````python
# Independent Track 2 interpretation declarations.

DECISION_SHAPE_DECLARATIONS = {
    "temperature": {
        "meaning": (
            "Celsius",
            "Fahrenheit",
        ),
        "structure": (
            "rounded up",
            "rounded down",
        ),
        "world": (
            "Baltic Sea",
            "Atlantic Ocean",
        ),
    },
    "location": {
        "meaning": (
            "location associated with the temperature reading",
        ),
        "structure": (
            "string",
        ),
        "world": (
            "corresponding international city",
            "corresponding U.S. city",
        ),
    },
    "date": {
        "meaning": (
            "date the reading was collected",
            "date the reported lowest temperature occurred",
        ),
        "structure": (
            "MM/DD/YYYY",
            "DD/MM/YYYY",
        ),
        "world": (
            "no alternate interpretation established",
        ),
    },
}


DECISION_SHAPE_CANDIDATES = {
    f"{data}.{facet}": interpretations
    for data, facets
    in DECISION_SHAPE_DECLARATIONS.items()
    for facet, interpretations
    in facets.items()
    if len(interpretations) > 1
}


DECISION_SHAPE_CANDIDATE_KEYS = tuple(
    DECISION_SHAPE_CANDIDATES
)


assert DECISION_SHAPE_CANDIDATE_KEYS == (
    "temperature.meaning",
    "temperature.structure",
    "temperature.world",
    "location.world",
    "date.meaning",
    "date.structure",
)

assert len(DECISION_SHAPE_CANDIDATES) == 6

assert all(
    len(interpretations) == 2
    for interpretations
    in DECISION_SHAPE_CANDIDATES.values()
)


DECISION_SHAPE_CANDIDATE_URIS = {
    key: candidate_uris(
        DECISION_SHAPE_DECLARATIONS,
        *key.split(".", 1),
    )
    for key in DECISION_SHAPE_CANDIDATE_KEYS
}

assert all(
    len(addresses) == 2
    for addresses
    in DECISION_SHAPE_CANDIDATE_URIS.values()
)

assert len({
    address
    for addresses
    in DECISION_SHAPE_CANDIDATE_URIS.values()
    for address in addresses
}) == 12


print_table({
    "Candidate Context Facet": list(
        DECISION_SHAPE_CANDIDATE_KEYS
    ),
    "Declared interpretations": [
        " / ".join(
            DECISION_SHAPE_CANDIDATES[key]
        )
        for key in DECISION_SHAPE_CANDIDATE_KEYS
    ],
    "Declared cgvu addresses": [
        " / ".join(
            DECISION_SHAPE_CANDIDATE_URIS[key]
        )
        for key in DECISION_SHAPE_CANDIDATE_KEYS
    ],
})

print(
    "Verified: 6 binary candidate facets "
    "define 2^6 = 64 complete states."
)
````

## **Materializing Every Complete Interpretation State**

A complete candidate state must affect the interpreted artifact before the Decision Rule is evaluated.

The experiment must not omit `location.world` or `date.meaning` merely because the current Decision Rule may not consume them. Their interpretations must remain visible in the materialized result so that the experiment can distinguish:

* a facet that does not participate in interpretation; from
* a facet that changes the interpreted context but is `Inert` with respect to this particular Verdict.

For each complete state, the materializer will:

1. use `date.structure` to select the row representing May;
2. interpret the selected temperature through `temperature.meaning`;
3. calculate its coldest plausible value through `temperature.structure`;
4. select the freezing threshold through `temperature.world`;
5. interpret the selected location through `location.world`;
6. identify the event represented by the date through `date.meaning`; and
7. apply the known Decision Rule to produce the Operational Verdict.

The resulting record will therefore preserve both:

* the complete interpreted context; and
* the smaller set of operational values actually consumed by the Decision Rule.

This separation is necessary to certify an `Inert Facet` without treating it as absent.


````python
# Complete Track 2 candidate-state materializer.
from itertools import product


DECISION_SHAPE_LOCATION_INTERPRETATIONS = {
    "corresponding international city": {
        "Paris": "Paris, France",
        "Cambridge": "Cambridge, England",
        "Melbourne": "Melbourne, Australia",
        "Athens": "Athens, Greece",
        "London": "London, England",
    },
    "corresponding U.S. city": {
        "Paris": "Paris, Texas",
        "Cambridge": "Cambridge, Massachusetts",
        "Melbourne": "Melbourne, Florida",
        "Athens": "Athens, Georgia",
        "London": "London, Kentucky",
    },
}


DECISION_SHAPE_STATES = tuple(
    product(
        *(
            DECISION_SHAPE_CANDIDATES[key]
            for key in DECISION_SHAPE_CANDIDATE_KEYS
        )
    )
)


DECISION_SHAPE_STATE_ID_BY_STATE = {
    state: f"DS{state_number:02d}"
    for state_number, state in enumerate(
        DECISION_SHAPE_STATES,
        1,
    )
}


assert len(DECISION_SHAPE_STATES) == 64
assert len(set(DECISION_SHAPE_STATES)) == 64

assert len({
    state_id
    for state_id
    in DECISION_SHAPE_STATE_ID_BY_STATE.values()
}) == 64


def materialize_decision_shape_state(state):
    """Materialize one complete six-candidate interpretation state."""
    interpretations = dict(
        zip(
            DECISION_SHAPE_CANDIDATE_KEYS,
            state,
        )
    )

    is_mm_dd = (
        interpretations["date.structure"]
        == "MM/DD/YYYY"
    )

    selected_row = select_row(
        db.get(1),
        QUERY_MONTH,
        is_mm_dd,
    )

    assert selected_row is not None

    raw_date = db.get(1)[
        "date"
    ][selected_row]

    raw_location = db.get(1)[
        "location"
    ][selected_row]

    raw_temperature = db.get(1)[
        "temperature"
    ][selected_row]

    interpreted_location = (
        DECISION_SHAPE_LOCATION_INTERPRETATIONS[
            interpretations["location.world"]
        ][raw_location]
    )

    interpreted_date = (
        f"{interpretations['date.meaning']}: "
        f"{raw_date}"
    )

    decision_trace = trace_decision(
        recorded_temperature=raw_temperature,
        is_celsius=(
            interpretations["temperature.meaning"]
            == "Celsius"
        ),
        is_rounded_up=(
            interpretations["temperature.structure"]
            == "rounded up"
        ),
        is_baltic=(
            interpretations["temperature.world"]
            == "Baltic Sea"
        ),
    )

    candidate_uri_assignments = {
        key: DECISION_SHAPE_CANDIDATE_URIS[
            key
        ][
            DECISION_SHAPE_CANDIDATES[
                key
            ].index(
                interpretations[key]
            )
        ]
        for key in DECISION_SHAPE_CANDIDATE_KEYS
    }

    return {
        "state identifier": (
            DECISION_SHAPE_STATE_ID_BY_STATE[
                state
            ]
        ),
        **interpretations,
        "selected row": selected_row + 1,
        "selected raw date": raw_date,
        "selected raw location": raw_location,
        "interpreted location": interpreted_location,
        "interpreted date": interpreted_date,
        **decision_trace,
        "candidate cgvu assignments": (
            candidate_uri_assignments
        ),
    }


DECISION_SHAPE_RESULTS = {
    state: materialize_decision_shape_state(
        state
    )
    for state in DECISION_SHAPE_STATES
}


assert len(DECISION_SHAPE_RESULTS) == 64

assert {
    result["state identifier"]
    for result in DECISION_SHAPE_RESULTS.values()
} == set(
    DECISION_SHAPE_STATE_ID_BY_STATE.values()
)

assert {
    result["location.world"]
    for result in DECISION_SHAPE_RESULTS.values()
} == set(
    DECISION_SHAPE_CANDIDATES[
        "location.world"
    ]
)

assert {
    result["date.meaning"]
    for result in DECISION_SHAPE_RESULTS.values()
} == set(
    DECISION_SHAPE_CANDIDATES[
        "date.meaning"
    ]
)

assert all(
    set(
        result["candidate cgvu assignments"]
    )
    == set(
        DECISION_SHAPE_CANDIDATE_KEYS
    )
    for result in DECISION_SHAPE_RESULTS.values()
)


print(
    "Verified: all 64 complete candidate states "
    "were independently materialized."
)

print(
    "Verified: location.world and date.meaning "
    "remain visible in every materialized result."
)
````

## **Certifying Essential and Inert Facets**

The complete truth table lets us perform a controlled test on each candidate facet.

A **comparison pair** consists of two states that:

* hold five candidate interpretations fixed;
* change only the one candidate being tested; and
* compare the two resulting Operational Verdicts.

Because each candidate has two interpretations, the 64 states form 32 unique pairs for each candidate:

$$
64 \div 2 = 32
$$

Across all six candidates, the experiment therefore checks:

$$
6 \times 32 = 192
$$

one-facet comparison pairs.

### **A Concrete Example: `temperature.meaning`**

Consider states `DS01` and `DS33`.

Both states use:

* the same raw temperature: `29`;
* `rounded up`;
* the `Baltic Sea` freezing threshold;
* the same location interpretation;
* the same date meaning; and
* `MM/DD/YYYY`.

Only `temperature.meaning` changes.

| State  | Interpretation of `29` | Coldest plausible temperature | Freezing threshold | Verdict |
| ------ | ---------------------- | ----------------------------: | -----------------: | ------- |
| `DS01` | Celsius                |                        83.7°F |             31.0°F | `GO`    |
| `DS33` | Fahrenheit             |                        28.5°F |             31.0°F | `NO-GO` |

The raw value and the other five candidate interpretations remain fixed, but changing Celsius to Fahrenheit changes the Verdict from `GO` to `NO-GO`.

This makes `DS01` and `DS33` a **pivotal pair** for `temperature.meaning`.

In plain language:

> A pair is pivotal when changing only the facet being tested is enough to change the decision.

### **What “Pivotal Pairs” Means**

Every candidate receives 32 controlled comparisons.

For example:

```text
temperature.meaning: 28 of 32 pivotal pairs
```

means:

* 32 Celsius-versus-Fahrenheit comparisons were tested;
* the other five interpretations were held fixed within every comparison; and
* 28 of those 32 changes reversed the Verdict.

Only one pivotal pair is required to prove that a candidate is `Essential`.

```text
one or more pivotal pairs
        │
        ▼
    Essential

zero pivotal pairs
        │
        ▼
      Inert
```

### **What “Pivotal Share” Means**

The **pivotal share** is the fraction of a candidate’s 32 comparison pairs that reverse the Verdict:

$$
\text{Pivotal Share} = \frac{\text{Pivotal Pairs}}{32}
$$

For `temperature.meaning`:

$$
\frac{28}{32}=87.5\%
$$

This means that changing Celsius to Fahrenheit reverses the Verdict in 87.5% of the declared settings of the other five candidates.

The pivotal share measures how broadly a facet influences this Decision Rule across the declared state space. It is not automatically a real-world probability because the experiment has not declared that these 64 states occur with equal probability.

### **How to Read the Classification Table**

Read each row in this order:

1. **Classification** answers whether the facet can ever change the Verdict.
2. **Pivotal pairs** reports how many of its 32 controlled comparisons changed the Verdict.
3. **Pivotal share** reports that count as a percentage of the 32 comparisons.

The expected insight is:

* `temperature.meaning` is `Essential` and influences the Verdict across most of the declared state space.
* `temperature.structure`, `temperature.world`, and `date.structure` are also `Essential`, but their effects occur only under particular combinations of the other interpretations.
* `location.world` and `date.meaning` change the interpreted context, but neither changes this particular `GO` or `NO-GO` Verdict in any of its 32 comparisons. They are therefore `Inert` for this Operational Decision.

`Inert` does not mean unimportant, meaningless, or safe to ignore in every situation. It means only that the facet cannot change this particular Verdict under this particular Decision Rule.

### **How to Read the Witness Table**

The second table displays one concrete pivotal pair for each Essential Facet.

For each row:

* `State A` and `State B` differ only in the named Essential Facet;
* `Interpretation A` and `Interpretation B` show the one change being tested; and
* `Verdict A` and `Verdict B` demonstrate that the change reverses the decision.

The witness table proves that each Essential Facet can matter. The classification table reports how often it matters across the complete declared state space.


````python
# Exact Essential/Inert certification over all one-facet pairs.

DECISION_SHAPE_COMPARISON_PAIRS = {}


for candidate_position, candidate_key in enumerate(
    DECISION_SHAPE_CANDIDATE_KEYS
):
    (
        first_interpretation,
        second_interpretation,
    ) = DECISION_SHAPE_CANDIDATES[
        candidate_key
    ]

    comparisons = []

    for state in DECISION_SHAPE_STATES:
        if (
            state[candidate_position]
            != first_interpretation
        ):
            continue

        changed_state = list(state)
        changed_state[candidate_position] = (
            second_interpretation
        )
        changed_state = tuple(changed_state)

        differing_positions = [
            position
            for position, (
                original_value,
                changed_value,
            ) in enumerate(
                zip(
                    state,
                    changed_state,
                )
            )
            if original_value != changed_value
        ]

        assert differing_positions == [
            candidate_position
        ]

        original_result = (
            DECISION_SHAPE_RESULTS[state]
        )

        changed_result = (
            DECISION_SHAPE_RESULTS[
                changed_state
            ]
        )

        original_verdict = original_result[
            "Operational Verdict"
        ]

        changed_verdict = changed_result[
            "Operational Verdict"
        ]

        comparisons.append({
            "candidate facet": candidate_key,
            "state A": state,
            "state B": changed_state,
            "state A identifier": (
                original_result[
                    "state identifier"
                ]
            ),
            "state B identifier": (
                changed_result[
                    "state identifier"
                ]
            ),
            "interpretation A": (
                first_interpretation
            ),
            "interpretation B": (
                second_interpretation
            ),
            "Verdict A": original_verdict,
            "Verdict B": changed_verdict,
            "pivotal": (
                original_verdict
                != changed_verdict
            ),
        })

    assert len(comparisons) == 32

    DECISION_SHAPE_COMPARISON_PAIRS[
        candidate_key
    ] = comparisons


assert sum(
    len(comparisons)
    for comparisons
    in DECISION_SHAPE_COMPARISON_PAIRS.values()
) == 192


DECISION_SHAPE_PIVOTAL_PAIR_COUNTS = {
    candidate_key: sum(
        comparison["pivotal"]
        for comparison in comparisons
    )
    for candidate_key, comparisons
    in DECISION_SHAPE_COMPARISON_PAIRS.items()
}


assert DECISION_SHAPE_PIVOTAL_PAIR_COUNTS == {
    "temperature.meaning": 28,
    "temperature.structure": 4,
    "temperature.world": 4,
    "location.world": 0,
    "date.meaning": 0,
    "date.structure": 4,
}


DECISION_SHAPE_CLASSIFICATION = {
    candidate_key: (
        "Essential"
        if pivotal_count > 0
        else "Inert"
    )
    for candidate_key, pivotal_count
    in DECISION_SHAPE_PIVOTAL_PAIR_COUNTS.items()
}


DECISION_SHAPE = tuple(
    candidate_key
    for candidate_key
    in DECISION_SHAPE_CANDIDATE_KEYS
    if (
        DECISION_SHAPE_CLASSIFICATION[
            candidate_key
        ]
        == "Essential"
    )
)


DECISION_SHAPE_INERT_FACETS = tuple(
    candidate_key
    for candidate_key
    in DECISION_SHAPE_CANDIDATE_KEYS
    if (
        DECISION_SHAPE_CLASSIFICATION[
            candidate_key
        ]
        == "Inert"
    )
)


assert DECISION_SHAPE == (
    "temperature.meaning",
    "temperature.structure",
    "temperature.world",
    "date.structure",
)

assert DECISION_SHAPE_INERT_FACETS == (
    "location.world",
    "date.meaning",
)


DECISION_SHAPE_VERDICT_REVERSAL_WITNESSES = {
    candidate_key: next(
        comparison
        for comparison in comparisons
        if comparison["pivotal"]
    )
    for candidate_key, comparisons
    in DECISION_SHAPE_COMPARISON_PAIRS.items()
    if any(
        comparison["pivotal"]
        for comparison in comparisons
    )
}


assert set(
    DECISION_SHAPE_VERDICT_REVERSAL_WITNESSES
) == set(
    DECISION_SHAPE
)


# The two Inert Facets still change the materialized context.
assert all(
    (
        DECISION_SHAPE_RESULTS[
            comparison["state A"]
        ]["interpreted location"]
        !=
        DECISION_SHAPE_RESULTS[
            comparison["state B"]
        ]["interpreted location"]
    )
    and not comparison["pivotal"]
    for comparison
    in DECISION_SHAPE_COMPARISON_PAIRS[
        "location.world"
    ]
)

assert all(
    (
        DECISION_SHAPE_RESULTS[
            comparison["state A"]
        ]["interpreted date"]
        !=
        DECISION_SHAPE_RESULTS[
            comparison["state B"]
        ]["interpreted date"]
    )
    and not comparison["pivotal"]
    for comparison
    in DECISION_SHAPE_COMPARISON_PAIRS[
        "date.meaning"
    ]
)


print(
    "Verified: 192 one-facet comparison pairs "
    "were evaluated."
)

print_table({
    "Candidate Context Facet": list(
        DECISION_SHAPE_CANDIDATE_KEYS
    ),
    "Classification": [
        DECISION_SHAPE_CLASSIFICATION[
            candidate_key
        ]
        for candidate_key
        in DECISION_SHAPE_CANDIDATE_KEYS
    ],
    "Pivotal pairs": [
        (
            f"{DECISION_SHAPE_PIVOTAL_PAIR_COUNTS[candidate_key]}"
            " of 32"
        )
        for candidate_key
        in DECISION_SHAPE_CANDIDATE_KEYS
    ],
    "Pivotal share": [
        (
            f"{DECISION_SHAPE_PIVOTAL_PAIR_COUNTS[candidate_key] / 32:.1%}"
        )
        for candidate_key
        in DECISION_SHAPE_CANDIDATE_KEYS
    ],
})


witnesses = list(
    DECISION_SHAPE_VERDICT_REVERSAL_WITNESSES.values()
)

print_table({
    "Essential Facet": [
        witness["candidate facet"]
        for witness in witnesses
    ],
    "State A": [
        witness["state A identifier"]
        for witness in witnesses
    ],
    "Interpretation A": [
        witness["interpretation A"]
        for witness in witnesses
    ],
    "Verdict A": [
        witness["Verdict A"]
        for witness in witnesses
    ],
    "State B": [
        witness["state B identifier"]
        for witness in witnesses
    ],
    "Interpretation B": [
        witness["interpretation B"]
        for witness in witnesses
    ],
    "Verdict B": [
        witness["Verdict B"]
        for witness in witnesses
    ],
})
````

## **The Decision Shape Is a Minimal 4-Junta**

A **(k)-junta** is a function whose output depends on no more than (k) of its possible inputs.

This experiment begins with six candidate inputs:

```text
temperature.meaning
temperature.structure
temperature.world
location.world
date.meaning
date.structure
```

The certification proves that the Operational Verdict depends on exactly four of them:

```text
temperature.meaning
temperature.structure
temperature.world
date.structure
```

Therefore, this Decision Rule is a **4-junta** over the six declared candidate facets.

The two remaining candidates are outside the Decision Shape:

```text
location.world
date.meaning
```

They still change the interpreted context, but changing either one while holding the other five candidates fixed never changes the Operational Verdict.

### **Why 64 States Reduce to 16 Decision States**

The complete experiment contains:

$$
2^6 = 64
$$

candidate interpretation states.

However, the two Inert Facets each have two interpretations. For every fixed assignment to the four Essential Facets, there are:

$$
2^2 = 4
$$

possible assignments to the two Inert Facets.

All four produce the same Verdict.

The 64 complete states can therefore be grouped into:

$$
64 \div 4 = 16
$$

decision-equivalent groups.

```text
64 complete interpretation states
              │
              ▼
ignore no context, but group states
that differ only in Inert Facets
              │
              ▼
16 decision-equivalent groups
```

This reduction does not remove the Inert Facets from the artifact. It shows that the current Decision Rule does not need them to determine its Verdict.

### **Why the 4-Junta Is Minimal**

The witness table contains at least one Verdict-reversal pair for each of the four Essential Facets.

That means none of the four can be removed while preserving the Decision Rule across the complete declared state space.

Therefore, the decision is not merely representable using four candidates. It requires all four:

$$
\boxed{
\text{Decision Shape} = \{
\texttt{temperature.meaning},\
\texttt{temperature.structure},\
\texttt{temperature.world},\
\texttt{date.structure}
\}
}
$$


The Decision Shape is consequently a **minimal 4-junta** for this Operational Decision.

# Track 3 — Context-Aware Join Proof

## A Join Is a Claim About Identity

A join combines rows from different tables. It does more than compare values: it decides which rows represent the same thing.

To perform a join, we select one or more fields called a **join key**. When the join-key values satisfy the comparison rule, the rows are paired.

For this experiment, the join key is:

```text
(location, date)
```

The intended rule is:

> Pair two rows when they describe the same location on the same real-world date.

The notebook's shared tables are:

### `table-1`

| temperature | location  | date       |
| ----------: | --------- | ---------- |
|          29 | Paris     | 05/04/2026 |
|          28 | Cambridge | 04/05/2026 |
|          30 | Melbourne | 02/06/2026 |
|          27 | Athens    | 11/07/2026 |
|          29 | London    | 01/12/2026 |

### `table-2`

| temperature | location   | date       |
| ----------: | ---------- | ---------- |
|          30 | Manchester | 05/04/2026 |
|          29 | Athens     | 02/06/2026 |
|          27 | Paris      | 04/05/2026 |
|          28 | Vancouver  | 09/03/2026 |
|          30 | London     | 01/12/2026 |

Only the Paris and London rows can pair under either join mode examined below; no other pair of rows shares a `(location, date)` key under either interpretation. This claim is executed and asserted in the pair-accuracy cell at the end of the track.

At first glance, joining these tables appears straightforward:

1. compare `location`;
2. compare `date`;
3. pair the rows when both values are equal.

The important question is what kind of equality the join is testing.

A conventional join tests whether the represented values are equal. It does not independently verify that those values carry the same `Meaning`, `Structure`, or `World` referent.

## The Context-Blind Join

Suppose we join the tables using ordinary raw-string equality:

```text
table-1.location == table-2.location
and
table-1.date == table-2.date
```

The London rows contain the same location string and the same date string:

```text
London
01/12/2026
```

The context-blind join therefore pairs them.

The Paris rows are not paired because their raw date strings differ:

```text
table-1: 05/04/2026
table-2: 04/05/2026
```

The join completes successfully. It produces a structurally valid result and raises no exception.

| Join mode                     | Paired location | Paired date | `table-1` temperature | `table-2` temperature |
| ----------------------------- | --------------- | ----------- | --------------------: | --------------------: |
| Context-blind raw-string join | London          | 01/12/2026  |                    29 |                    30 |

Nothing in this output announces that the join may have paired the wrong rows.

Raw-string equality has silently acted as an undeclared interpretation of the two Dark `date.structure` facets: it substitutes the assumption "identical characters represent the same date" for declared contextual equality.


## Why the Successful Join Is Wrong

The raw date values do not carry their interpretation inside the strings.

For this finite experiment, the source-declared date structures are:

```text
table-1.date.structure = MM/DD/YYYY
table-2.date.structure = DD/MM/YYYY
```

Applying those declarations changes what the strings represent.

### Paris

```text
table-1: 05/04/2026 interpreted as MM/DD/YYYY
         = May 4, 2026

table-2: 04/05/2026 interpreted as DD/MM/YYYY
         = May 4, 2026
```

The Paris strings are syntactically different but contextually equal.

They represent the same real-world date and should be joined.

### London

```text
table-1: 01/12/2026 interpreted as MM/DD/YYYY
         = January 12, 2026

table-2: 01/12/2026 interpreted as DD/MM/YYYY
         = December 1, 2026
```

The London strings are syntactically equal but contextually different.

They represent different real-world dates and should not be joined.

The context-blind join therefore makes two errors:

1. it produces a false London pairing; and
2. it misses the true Paris pairing.

> A successful join proves that values satisfied a comparison rule. It does not prove that the rows represent the same real-world thing.

### Why This Matters for AI-Assisted Data Work

Users can upload CSV files and ask an AI to answer questions across them.

Answering the question may require the AI to generate SQL, Python, or another data transformation containing a join. If the join keys' interpretations have not been verified, the generated code can run successfully while silently selecting the wrong evidence.

The danger is not that the join crashes.

The danger is that it does not crash.

A structurally valid result can be mistaken for a contextually valid result.


## **What the Context Graph Protocol Adds**

A join should not begin by assuming that matching characters represent matching entities.

It should first determine whether the join-key fields are contextually compatible.

Every data field has three `Context Facet`s:

```text
Meaning
Structure
World
```

For this controlled experiment, the relevant declarations are hard-coded as source-provided metadata. CGP is not being credited with independently discovering these declarations. Both join modes below have the same declarations available; the modes differ only in whether the declarations are applied. That is what makes this a controlled comparison of the join behavior rather than a comparison of two different information states.

### Join-Key Context Declarations

| Join key   | Source    | `Meaning`                               | `Structure`      | `World`                    |
| ---------- | --------- | --------------------------------------- | ---------------- | -------------------------- |
| `location` | `table-1` | Location of the temperature observation | City-name string | Declared geographic entity |
| `location` | `table-2` | Location of the temperature observation | City-name string | Declared geographic entity |
| `date`     | `table-1` | Date of the temperature observation     | MM/DD/YYYY       | Gregorian civil date       |
| `date`     | `table-2` | Date of the temperature observation     | DD/MM/YYYY       | Gregorian civil date       |

Only the fields participating in the join key are evaluated here.

The `temperature` fields become important after the join, when the joined evidence enters an `Operational Decision`. Their Context Facets do not determine whether two rows represent the same location and date.

### CGP Responsibilities at the Join Boundary

The protocol separates the join process into explicit responsibilities:

1. The `Observatron` observes an attempted join on `location` and `date`.
2. The `Intent Map` declares which join-key Context Facets must be checked and which `Decision Route`s are available.
3. The `Context State` records whether each required facet is `LIT` or `DARK`.
4. The `Governor` evaluates the Context State against the Intent Map and returns a Governor `Verdict` — `ACT`, `ASK`, or `HALT` — together with a selected `Decision Route`.
5. The registered `Handler` performs the routed behavior: joining directly, canonicalizing before the join, requesting a missing declaration, or refusing the join and recording the conflict.

CGP does not require the two sources to use identical representations.

It requires sufficient declared context to determine whether the representations are compatible.


## Governing the Join

The Governor's Verdict vocabulary is the one declared in the notebook's component table: `ACT`, `ASK`, or `HALT`. What differs between join conditions is not the Verdict alphabet but the selected `Decision Route` — the registered Handler behavior the Verdict activates.

| Context condition                                                    | Governor Verdict | Routed Handler behavior                          |
| -------------------------------------------------------------------- | ---------------- | ------------------------------------------------ |
| The required facets are `LIT` and their declared referents match     | `ACT`            | Join directly on the declared keys               |
| The structures differ, but a declared lossless transformation exists | `ACT`            | Canonicalize both keys, then join                |
| A required facet remains `DARK`                                      | `ASK`            | Request the missing declaration; the join waits  |
| The declared Meanings or World referents conflict                    | `HALT`           | Refuse the join and record the conflict          |

The four registered Decision Routes are:

```text
cgp:/track-3/handlers/join-on-declared-keys
cgp:/track-3/handlers/canonicalize-then-join
cgp:/track-3/handlers/ask-join-key-declaration
cgp:/track-3/handlers/refuse-join-record-conflict
```

For this experiment, the declared join-key evaluation is:

| Join-key check       | Declared comparison result                            |
| -------------------- | ----------------------------------------------------- |
| `location.meaning`   | `MATCH`                                               |
| `location.structure` | `MATCH`                                               |
| `location.world`     | `MATCH`                                               |
| `date.meaning`       | `MATCH`                                               |
| `date.structure`     | `NO MATCH` — declared lossless canonicalization exists |
| `date.world`         | `MATCH`                                               |
| Governor Verdict     | `ACT`                                                 |
| Decision Route       | `cgp:/track-3/handlers/canonicalize-then-join`        |

The date structures are not equivalent:

```text
MM/DD/YYYY != DD/MM/YYYY
```

In plain language, "declared lossless canonicalization exists" means this difference is safe to fix:

- **Declared** — each source has stated exactly which date format it uses, so the conversion rule comes from recorded source context, not from a guess the join makes on its own.
- **Lossless** — rewriting a date from one declared format into another still names the exact same day; nothing is approximated or thrown away.
- **Canonicalization** — both sides are rewritten into one shared standard form before they are compared:

```text
YYYY-MM-DD
```

Both sources declare Gregorian civil dates, so this shared form exists for every value. Only after that transformation does the join compare the keys.

The next cell executes this evaluation with a minimal join-boundary Governor. Its declared route-selection precedence is: referent conflicts `HALT`, Dark facets `ASK`, transformable Structure differences canonicalize, and fully matching keys join directly. The cell asserts that the executed evaluation reproduces the table above.

````python
# Exact join-boundary Governor routing proof.

# The declared join-key Context Facet referents for each source.
# A value of None declares that the facet is DARK: no resolving
# referent is carried for that crossing.
TRACK3_JOIN_KEY_DECLARATIONS = {
    "location": {
        "meaning": {
            "table-1": "Location of the temperature observation",
            "table-2": "Location of the temperature observation",
        },
        "structure": {
            "table-1": "City-name string",
            "table-2": "City-name string",
        },
        "world": {
            "table-1": "Declared geographic entity",
            "table-2": "Declared geographic entity",
        },
    },
    "date": {
        "meaning": {
            "table-1": "Date of the temperature observation",
            "table-2": "Date of the temperature observation",
        },
        "structure": {
            "table-1": "MM/DD/YYYY",
            "table-2": "DD/MM/YYYY",
        },
        "world": {
            "table-1": "Gregorian civil date",
            "table-2": "Gregorian civil date",
        },
    },
}

# The declared lossless canonicalizations available to Handlers.
# Only Structure differences listed here are transformable.
TRACK3_DECLARED_CANONICALIZATIONS = {
    ("MM/DD/YYYY", "DD/MM/YYYY"): "YYYY-MM-DD",
    ("DD/MM/YYYY", "MM/DD/YYYY"): "YYYY-MM-DD",
}

# The Decision Routes registered for the join boundary.
TRACK3_JOIN_ROUTES = {
    "direct": "cgp:/track-3/handlers/join-on-declared-keys",
    "canonicalize": "cgp:/track-3/handlers/canonicalize-then-join",
    "ask": "cgp:/track-3/handlers/ask-join-key-declaration",
    "refuse": "cgp:/track-3/handlers/refuse-join-record-conflict",
}


def track3_governor(
    join_key_declarations,
    canonicalizations,
):
    """Evaluate the declared join-key facets and select a route.

    The Verdict vocabulary is the notebook's declared Governor
    alphabet: ACT, ASK, or HALT. The declared route-selection
    precedence is: referent conflicts HALT, Dark facets ASK,
    transformable Structure differences canonicalize, and fully
    matching keys join directly.
    """
    checks = {}
    dark_crossings = []
    conflicting_crossings = []
    transformable_crossings = []

    for join_key, facets in join_key_declarations.items():
        for facet_name, declared in facets.items():
            crossing = f"{join_key}.{facet_name}"
            left = declared["table-1"]
            right = declared["table-2"]

            if left is None or right is None:
                checks[crossing] = "DARK"
                dark_crossings.append(crossing)
            elif left == right:
                checks[crossing] = "MATCH"
            elif (
                facet_name == "structure"
                and (left, right) in canonicalizations
            ):
                checks[crossing] = "NO MATCH"
                transformable_crossings.append(crossing)
            else:
                checks[crossing] = "NO MATCH"
                conflicting_crossings.append(crossing)

    if conflicting_crossings:
        instruction = {
            "Governor Verdict": "HALT",
            "Decision Route": TRACK3_JOIN_ROUTES["refuse"],
        }
    elif dark_crossings:
        instruction = {
            "Governor Verdict": "ASK",
            "Decision Route": TRACK3_JOIN_ROUTES["ask"],
        }
    elif transformable_crossings:
        instruction = {
            "Governor Verdict": "ACT",
            "Decision Route": TRACK3_JOIN_ROUTES["canonicalize"],
        }
    else:
        instruction = {
            "Governor Verdict": "ACT",
            "Decision Route": TRACK3_JOIN_ROUTES["direct"],
        }

    condition_sets = {
        "dark": tuple(dark_crossings),
        "conflicting": tuple(conflicting_crossings),
        "transformable": tuple(transformable_crossings),
    }

    return instruction, checks, condition_sets


(
    track3_instruction,
    track3_checks,
    track3_condition_sets,
) = track3_governor(
    TRACK3_JOIN_KEY_DECLARATIONS,
    TRACK3_DECLARED_CANONICALIZATIONS,
)


# The executed evaluation must reproduce the declared join-key evaluation table.
assert track3_checks == {
    "location.meaning": "MATCH",
    "location.structure": "MATCH",
    "location.world": "MATCH",
    "date.meaning": "MATCH",
    "date.structure": "NO MATCH",
    "date.world": "MATCH",
}

# The only non-matching crossing is transformable, not conflicting
# and not DARK.
assert track3_condition_sets["conflicting"] == ()
assert track3_condition_sets["dark"] == ()
assert track3_condition_sets["transformable"] == (
    "date.structure",
)

# The Governor's Verdict uses the declared alphabet and routes
# through canonicalization.
assert track3_instruction == {
    "Governor Verdict": "ACT",
    "Decision Route": TRACK3_JOIN_ROUTES["canonicalize"],
}


print_table({
    "Join-key check": list(track3_checks),
    "Comparison result": list(track3_checks.values()),
})

print_table({
    "Governor Verdict": [
        track3_instruction["Governor Verdict"]
    ],
    "Decision Route": [
        track3_instruction["Decision Route"]
    ],
})

print(
    "Verified: the executed join-boundary evaluation reproduces "
    "the declared table, and the Governor returns ACT with the "
    "canonicalize-then-join Decision Route."
)
````

## Additional Controlled Join Conditions

The Paris/London witness demonstrates one structure mismatch.

A broader finite test should also distinguish other conditions that may occur at a join boundary. Each condition below is constructed as a declared fixture and executed against the same join-boundary Governor; the next cell asserts every expected routing.

| Controlled condition                                                  | Context-blind default               | Governor Verdict | Routed behavior                              |
| --------------------------------------------------------------------- | ----------------------------------- | ---------------- | ----------------------------------------------------- |
| Meaning, Structure, and World match                                   | Proceed                             | `ACT`                     | Join directly on the declared keys                    |
| Date structures differ but both are declared and transformable        | Proceed using raw values            | `ACT`                     | Canonicalize both keys, then join                     |
| One source's date structure is unknown                                | Proceed using an undeclared default | `ASK`                     | Request the missing declaration; the join waits       |
| One date means collection date and the other means publication date   | Proceed if values match             | `HALT`                    | Refuse the join; record the Meaning conflict          |
| One location denotes Paris, France and the other denotes Paris, Texas | Proceed if labels match             | `HALT`                    | Refuse the join; record the World conflict            |
| A non-key temperature unit remains unresolved                         | Proceed                             | `ACT`                     | Join; the downstream facet remains `DARK`             |

A Governor does not need all of these routings. A perfectly valid Intent Map can declare one conservative rule:

```text
If any Context Facet differs, HALT.
```

That blunt rule never produces a false pairing, and it needs no information beyond the facet comparisons themselves. **Its cost is precision, not correctness**: under this experiment's declarations it would refuse the join outright — forfeiting the true Paris pairing — because it has no way to treat a fixable format difference differently from a real conflict. The next cell executes this rule alongside the context-aware Governor and asserts both results.

Each richer routing is purchased with a specific piece of declared context, which the Intent Map may or may not contain:

- `ACT`, routed through canonicalization — in plain terms: proceed with the join, but first run both keys through the declared conversion so they are compared in one shared format. This routing exists only because both date formats are declared; without those declarations there is no safe conversion to route through.
- `ASK` — available only when the Intent Map declares where a missing declaration can come from, so the Governor can request it instead of guessing or refusing.
- `HALT` — under the richer Intent Map, reserved for declared Meaning or World conflicts, where no conversion can make equal-looking values represent the same thing.

And an ambiguity outside the join key may allow the join to proceed while still preventing a downstream Operational Decision — the join-boundary check and Track 2's Decision Shape check remain separate gates.

A Governor limited to the blunt rule is not wrong; it is operating at the floor of its declared context. Each added declaration — a format, a conversion, an answer source — lets it distinguish one more of these conditions and route it more precisely.

## Measuring the Dark Join State

Before the date structures are resolved, each source has two declared candidate interpretations:

```text
MM/DD/YYYY
DD/MM/YYYY
```

The two unresolved facets therefore define

$$
2 \times 2 = 4
$$

admissible joint interpretation states, with Hartley uncertainty

$$
H_0 = \log_2(4) = 2 \text{ bits}
$$

After both source structures are declared, one joint interpretation state remains:

$$
H_0 = \log_2(1) = 0 \text{ bits}
$$

For this witness, resolving the two join-key facets reduces the declared interpretation set from four states to one — from two bits to zero.

This measurement records only the size of the finite unresolved interpretation set. It does not assign probabilities to the candidate states. An expected-cost claim would additionally require declared probabilities over the unresolved interpretation states, which this proof does not introduce.

### What the Cells Assert

Track 3's claims are executed across four cells.

````python
from datetime import datetime
from math import log2


# Each Dark date.structure facet has two declared
# candidate interpretations.
DATE_STRUCTURE_CANDIDATES = {
    "table-1": (
        "MM/DD/YYYY",
        "DD/MM/YYYY",
    ),
    "table-2": (
        "MM/DD/YYYY",
        "DD/MM/YYYY",
    ),
}

# The source-declared structures used by the
# Resolved/LIT join.
DATE_STRUCTURES = {
    "table-1": "%m/%d/%Y",
    "table-2": "%d/%m/%Y",
}


# Measure the finite uncertainty set without assigning
# probabilities to its members.
dark_join_state_count = (
    len(DATE_STRUCTURE_CANDIDATES["table-1"])
    * len(DATE_STRUCTURE_CANDIDATES["table-2"])
)

dark_join_uncertainty_bits = log2(
    dark_join_state_count
)

# Resolving both facets leaves one admissible state.
resolved_join_state_count = 1

resolved_join_uncertainty_bits = log2(
    resolved_join_state_count
)

assert dark_join_state_count == 4
assert dark_join_uncertainty_bits == 2.0

assert resolved_join_state_count == 1
assert resolved_join_uncertainty_bits == 0.0

assert (
    resolved_join_state_count
    < dark_join_state_count
)

assert (
    resolved_join_uncertainty_bits
    < dark_join_uncertainty_bits
)


def canonical_date(value, structure):
    """Interpret a declared date and return ISO form."""
    return datetime.strptime(
        value,
        structure,
    ).date().isoformat()


def join_on_location_date(
    left,
    right,
    left_date_key,
    right_date_key,
):
    """Inner join on interpreted (location, date) keys."""
    rows = []

    for i, (left_location, left_date) in enumerate(
        zip(left["location"], left["date"])
    ):
        left_key = (
            left_location,
            left_date_key(left_date),
        )

        for j, (right_location, right_date) in enumerate(
            zip(right["location"], right["date"])
        ):
            right_key = (
                right_location,
                right_date_key(right_date),
            )

            if left_key == right_key:
                rows.append(
                    (
                        left_location,
                        left_key[1],
                        left["temperature"][i],
                        right["temperature"][j],
                    )
                )

    return rows


# Resolved/LIT mode:
# interpret each date with its declared structure,
# canonicalize it, and then test equality.
resolved_join = join_on_location_date(
    db.get(1),
    db.get(2),
    lambda value: canonical_date(
        value,
        DATE_STRUCTURES["table-1"],
    ),
    lambda value: canonical_date(
        value,
        DATE_STRUCTURES["table-2"],
    ),
)


# Context-blind mode:
# leave both date structures unresolved and let raw-string
# equality act as the undeclared default.
context_blind_join = join_on_location_date(
    db.get(1),
    db.get(2),
    lambda value: value,
    lambda value: value,
)


expected_resolved = [
    (
        "Paris",
        "2026-05-04",
        29,
        27,
    )
]

expected_context_blind = [
    (
        "London",
        "01/12/2026",
        29,
        30,
    )
]


# Reaching these assertions proves that both joins
# executed without an exception.
assert isinstance(resolved_join, list)
assert isinstance(context_blind_join, list)

assert resolved_join == expected_resolved
assert context_blind_join == expected_context_blind

assert resolved_join != context_blind_join


# The context-blind join misses the Paris pairing
# produced by the declared date structures.
assert not any(
    (
        location,
        temperature_1,
        temperature_2,
    )
    == (
        "Paris",
        29,
        27,
    )
    for (
        location,
        _date,
        temperature_1,
        temperature_2,
    ) in context_blind_join
)


# Hold every downstream temperature Context Facet fixed.
# Only the join-key date.structure facets are varied.
JOIN_DECISION_CONTEXT = {
    "is_celsius": False,
    "is_rounded_up": False,
    "is_baltic": False,
}


# The Operational Decision consumes the table-2
# temperature selected by each join.
resolved_decision_temperature = (
    resolved_join[0][3]
)

context_blind_decision_temperature = (
    context_blind_join[0][3]
)


resolved_verdict = decide(
    resolved_decision_temperature,
    **JOIN_DECISION_CONTEXT,
)

context_blind_verdict = decide(
    context_blind_decision_temperature,
    **JOIN_DECISION_CONTEXT,
)


assert resolved_decision_temperature == 27
assert context_blind_decision_temperature == 30

assert resolved_verdict == "NO-GO"
assert context_blind_verdict == "GO"

assert resolved_verdict != context_blind_verdict


# Treat the Resolved/LIT Verdict as the recorded
# source-intent Verdict for this finite witness.
source_intent_verdict = resolved_verdict


(
    resolved_realized_cash,
    resolved_optimal_cash,
    resolved_regret,
) = regret_outcome(
    resolved_verdict,
    source_intent_verdict,
)


(
    context_blind_realized_cash,
    context_blind_optimal_cash,
    context_blind_regret,
) = regret_outcome(
    context_blind_verdict,
    source_intent_verdict,
)


assert (
    resolved_realized_cash,
    resolved_optimal_cash,
    resolved_regret,
) == (
    0,
    0,
    0,
)

assert (
    context_blind_realized_cash,
    context_blind_optimal_cash,
    context_blind_regret,
) == (
    -2_000_000,
    0,
    2_000_000,
)


comparison = {
    "join mode": [
        "Resolved/LIT",
        "Context-blind raw-string default",
    ],
    "Context state": [
        "LIT",
        "DARK",
    ],
    "admissible date-structure states": [
        resolved_join_state_count,
        dark_join_state_count,
    ],
    "Hartley uncertainty (bits)": [
        resolved_join_uncertainty_bits,
        dark_join_uncertainty_bits,
    ],
    "location": [
        resolved_join[0][0],
        context_blind_join[0][0],
    ],
    "date": [
        resolved_join[0][1],
        context_blind_join[0][1],
    ],
    "table-1 temperature": [
        resolved_join[0][2],
        context_blind_join[0][2],
    ],
    "table-2 temperature": [
        resolved_join[0][3],
        context_blind_join[0][3],
    ],
    "Operational Verdict": [
        resolved_verdict,
        context_blind_verdict,
    ],
    "realized cash": [
        "$0",
        "-$2,000,000",
    ],
    "regret": [
        "$0",
        "$2,000,000",
    ],
}


print_table(comparison)

print(
    "Verified: resolving both join-key Context Facets "
    "reduces the admissible state count from "
    f"{dark_join_state_count} to "
    f"{resolved_join_state_count} and the Hartley "
    "uncertainty from "
    f"{dark_join_uncertainty_bits:.0f} bits to "
    f"{resolved_join_uncertainty_bits:.0f} bits."
)

print(
    "Verified: the context-blind join executes without "
    "an error, selects different evidence, reverses the "
    "Operational Verdict, and produces "
    f"${context_blind_regret:,} of regret under the "
    "notebook's declared payoff ledger."
)
````

### **The Decidatron: Many Observatrons, One Gauge**

A **Decidatron** registers many Observatrons — one per boundary, one per captain — and measures the state of the network: **coverage**, the fraction of Context Facets that are Lit, per boundary and in total. It commits no interpretation of any facet. But like every node, it carries its own decision model: a Decidatron **routes with its Governor on the state of the network** — a collective brain of the many brains. Same outer loop, same rule: `HALT` / `ASK` / `ACT` are declarations read from its Intent Map — a network-level Intent Map whose rules are about coverage, not spikes — and the meaning of every handler lives at its URI. That is how a declared rule like *coverage below one half* dispatches an `ASK` such as *"less than half of this network's facets are documented."* Below: three captains with three documentation habits, one network report, and the Decidatron's Governor routing on it. The Admiral's economics — the same fleets, with regret drawn as a bar — arrive in Track 4-C, after the expected-regret lattice they price against is built.

````python
# Preview - the Decidatron: many Observatrons, one gauge, its own Governor.
# Capability 1: the GLOBAL ROUTER REGISTRY. Coverage is the scale variable,
# measured in the loop's standard unit: crossings.


def hbar(frac, color="#1a7f37", width=220):
    """A one-line HTML bar. Presentation, not data."""
    return Raw(f'<div style="background:#eee;width:{width}px;height:14px;'
               f'border-radius:3px;overflow:hidden">'
               f'<div style="background:{color};width:{int(width*frac)}px;'
               f'height:14px"></div></div>')

NETWORK_INTENT_MAP = {
    "rules": [
        {"if":   {"coverage-below": 0.5},
         "then": {"verdict": "ASK",
                  "handler": "cgvu:/network/handlers/document-facets"},
         "else": {"verdict": "ACT",
                  "handler": "cgvu:/network/handlers/log-coverage"}},
    ],
}

class Decidatron:
    """A collective brain of the many brains. The Decidatron is a network
    node with capabilities, and alpha instantiates its first: the GLOBAL
    ROUTER REGISTRY - the shared vocabulary of cgvu address objects, held
    live in memory, that routing across the network resolves against.
    Alongside it, the node measures the state of the network (coverage) and
    routes on that state with its own Governor - a decision model for
    routing. It commits no interpretation of any facet: the decision is the
    route; the semantics are not - verdicts are declarations read from its
    Intent Map, and the meaning of every handler lives at its URI.
    (Single-rule alpha; rule iteration with HALT absorbing, and further
    capabilities, arrive with Beta.)"""

    def __init__(self, name, intent_map, registry=()):
        self.name, self.intent_map = name, intent_map
        self.registry, self.members = tuple(registry), []

    def on_vocabulary(self):
        """Global Router Registry integrity: every address a member holds
        resolves against the registry this Decidatron is home to -
        resolve_facet's off-space check, lifted to network scale."""
        held = {uri for obs in self.members for uri in obs.facets.values() if uri}
        return held <= set(self.registry), held - set(self.registry)

    def crossing_coverage(self, obs):
        """Coverage in the loop's standard unit - crossings, not facet
        places ("Why crossings ... is the standard"). Consumes the Facet
        Crossings Calculator's table X and the Observatron's state API:
        a crossing is Lit when the member's facet behind it is Lit."""
        lit = sum(1 for sp, f in zip(X["spike"], X["facet"])
                  if obs.state(sp, f) == "LIT")
        return lit, len(X["index"])

    def coverage(self):
        lit = tot = 0
        for obs in self.members:
            l, t = self.crossing_coverage(obs)
            lit += l; tot += t
        return lit, tot

    def register(self, obs):
        assert all((sp, f) in obs.facets for sp, f in zip(X["spike"], X["facet"])), \
            "member's spikes must align with the loop's crossings"
        self.members.append(obs)
        return len(self.members)

    def report(self):
        rows = {"boundary": [], "crossings": [], "lit": [], "coverage": [], "gauge": []}
        for obs in self.members:
            l, t = self.crossing_coverage(obs)
            rows["boundary"].append(obs.name)
            rows["crossings"].append(t)
            rows["lit"].append(l)
            rows["coverage"].append(f"{l / t:.0%}")
            rows["gauge"].append(hbar(l / t))
        lit, tot = self.coverage()
        rows["boundary"].append(f"NETWORK: {self.name}")
        rows["crossings"].append(tot); rows["lit"].append(lit)
        rows["coverage"].append(f"{lit / tot:.0%}")
        rows["gauge"].append(hbar(lit / tot))
        return rows

    def governor_verdict(self):
        """The network's decision model for routing: decides the route from
        the declared responses for the measured coverage. The decision is the
        route; the semantics are not - meaning stays at the handler URIs."""
        lit, tot = self.coverage()
        rule = self.intent_map["rules"][0]
        branch = "then" if lit / tot < rule["if"]["coverage-below"] else "else"
        return rule[branch]["verdict"], rule[branch]["handler"], lit, tot

# three captains, three boundaries, three documentation habits
fleet_obs = {}
for captain in ("captain-alpha", "captain-beta", "captain-gamma"):
    obs = Observatron(
        captain,
        LOOP_INTENT_MAP,
    )
    for artifact, columns in LOOP_ARTIFACTS.items():
        obs.observe(artifact, columns)
    fleet_obs[captain] = obs

for spike in fleet_obs["captain-beta"].spikes:
    resolve_facet(
        fleet_obs["captain-beta"],
        spike["spike"],
        "meaning",
        candidate_uris(
            LOOP_CHOICES,
            spike["data"],
            "meaning",
        )[0],
    )

for spike in fleet_obs["captain-gamma"].spikes:
    for facet in ("meaning", "structure"):
        resolve_facet(
            fleet_obs["captain-gamma"],
            spike["spike"],
            facet,
            candidate_uris(
                LOOP_CHOICES,
                spike["data"],
                facet,
            )[0],
        )

# capability 1: the global router registry - the facet shards AND the handlers
# the network Governor routes to. One address book for all routing.
network_handlers = [b["handler"] for r in NETWORK_INTENT_MAP["rules"]
                    for b in (r["then"], r["else"])]

decidatron_1 = Decidatron(
    "decidatron-1",
    NETWORK_INTENT_MAP,
    registry=LOOP_URIS + network_handlers,
)

for obs in fleet_obs.values():
    decidatron_1.register(obs)

ok, strays = decidatron_1.on_vocabulary()
assert ok, f"off-registry addresses in the network: {sorted(strays)}"
print(
    f"capability 1 - global router registry: {decidatron_1.name} holds "
    f"{len(decidatron_1.registry)} addresses "
    f"({len(LOOP_URIS)} facet shards + "
    f"{len(network_handlers)} handlers); "
    f"every Lit facet is on-vocabulary"
)
print_table(decidatron_1.report())

# the routes file: declarations above, programmable logic on the same page.
# Each handler is deterministic code registered at its address.
def document_facets_handler(lit, tot):
    print(f"DECIDATRON ADVISORY: {lit}/{tot} crossings are Lit "
          f"({lit / tot:.0%}) - less than half of this network's crossings "
          f"are covered by documented facets, let alone woven into prompts "
          f"or UX.")

def log_coverage_handler(lit, tot):
    print(f"coverage log: {lit}/{tot} crossings Lit ({lit / tot:.0%})")

HANDLERS = {
    "cgvu:/network/handlers/document-facets": document_facets_handler,
    "cgvu:/network/handlers/log-coverage":    log_coverage_handler,
}
assert set(HANDLERS) == set(network_handlers), "routes and code must cover each other"

verdict, handler, lit, tot = decidatron_1.governor_verdict()
assert handler in decidatron_1.registry, "the verdict's handler must itself be on-registry"
print(f"decidatron-1 governor verdict: {verdict} -> {handler}")
HANDLERS[handler](lit, tot)   # dispatch: the code lives at the address

print(
    "Verified: three boundaries registered; every Lit facet is on-vocabulary; "
    "the network Governor routed on measured coverage through its declared "
    "Intent Map."
)
````

````python
# Exact shared-resolution-record proof.

TRACK4_COHORTS = {
    "captain-alpha": "north-fleet",
    "captain-beta": "north-fleet",
    "captain-gamma": "north-fleet",
}


def resolution_record(decidatron, cohorts):
    """One row per bound facet across the network's members."""
    rows = []

    for obs in decidatron.members:
        for record in obs.spikes:
            for facet in LOOP_CONTEXT_FACETS:
                uri = obs.uri(record["spike"], facet)

                if uri:
                    rows.append({
                        "member": obs.name,
                        "artifact": record["artifact"],
                        "data": record["data"],
                        "facet": facet,
                        "address": uri,
                        "cohort": cohorts[obs.name],
                    })

    return rows


track4_record = resolution_record(decidatron_1, TRACK4_COHORTS)

track4_rows_by_member = {
    name: sum(1 for row in track4_record if row["member"] == name)
    for name in TRACK4_COHORTS
}

# Alpha documents nothing; beta resolved meaning at every crossing;
# gamma resolved meaning and structure at every crossing.
assert track4_rows_by_member["captain-alpha"] == 0
assert track4_rows_by_member["captain-beta"] == len(
    fleet_obs["captain-beta"].spikes
)
assert track4_rows_by_member["captain-gamma"] == 2 * len(
    fleet_obs["captain-gamma"].spikes
)

assert len(fleet_obs["captain-beta"].spikes) == 12
assert len(track4_record) == 12 + 24

# Every recorded address is on the network's declared vocabulary.
assert all(
    row["address"] in decidatron_1.registry
    for row in track4_record
)

print_table({
    "member": list(track4_rows_by_member),
    "cohort": [
        TRACK4_COHORTS[name]
        for name in track4_rows_by_member
    ],
    "recorded resolutions": list(
        track4_rows_by_member.values()
    ),
})

print(
    f"Verified: the resolution record holds {len(track4_record)} rows "
    "(alpha 0, beta 12, gamma 24), every address on-vocabulary, every "
    "row tagged with its member's declared cohort."
)
````

### **The Cohort Suggestion (Collaborative Filtering, Minimal)**

The record supports the smallest useful form of collaborative filtering:

| Recommender concept | CGP realization |
| --- | --- |
| users | network members (captains) |
| items | Context Facet crossings |
| ratings | recorded resolving addresses |
| recommendation | a *suggestion*: a prefill for the `ASK` route |
| cold start | a new member with only a cohort attribute |

A new member, **captain-delta**, joins `north-fleet` with every facet `DARK`. At the crossing (`table-1` temperature, `meaning`), the network computes the recorded cohort frequency — 2 of 2 recorded resolutions are the Celsius address — and emits a suggestion with that exact support.

The design rule that keeps the gauge honest: **a suggestion is not a resolution.** It attaches to the member's `ASK` route as a prefill. The Context state stays `DARK`, the Governor's verdict stays `ASK`, and no member's bound facets change. The cell asserts all three.

````python
# Exact cohort-suggestion proof.
from fractions import Fraction

TRACK4_SUGGESTION_CROSSING = ("table-1", "temperature", "meaning")


def cohort_suggestion(records, cohort, crossing):
    """The recorded cohort frequency at one crossing, as a suggestion."""
    artifact, data, facet = crossing

    votes = [
        row["address"]
        for row in records
        if row["cohort"] == cohort
        and row["artifact"] == artifact
        and row["data"] == data
        and row["facet"] == facet
    ]

    assert votes, "no recorded resolutions at this crossing in this cohort"

    top = max(set(votes), key=votes.count)

    return {
        "address": top,
        "support": Fraction(votes.count(top), len(votes)),
        "recorded": len(votes),
    }


# captain-delta joins the network and the cohort, fully DARK.
delta_obs = Observatron("captain-delta", LOOP_INTENT_MAP)

for artifact, columns in LOOP_ARTIFACTS.items():
    delta_obs.observe(artifact, columns)

TRACK4_COHORTS["captain-delta"] = "north-fleet"
decidatron_1.register(delta_obs)

assert len(decidatron_1.members) == 4

TRACK4_DELTA_SPIKE = next(
    record["spike"]
    for record in delta_obs.spikes
    if record["artifact"] == "table-1"
    and record["data"] == "temperature"
)

# Snapshot every member's bound facets before the suggestion path runs.
track4_facets_before = {
    (obs.name, place): uri
    for obs in decidatron_1.members
    for place, uri in obs.facets.items()
}

track4_suggestion = cohort_suggestion(
    track4_record,
    "north-fleet",
    TRACK4_SUGGESTION_CROSSING,
)

CELSIUS_URI, FAHRENHEIT_URI = candidate_uris(
    LOOP_CHOICES,
    "temperature",
    "meaning",
)

# The suggestion is the exact recorded frequency: 2 of 2, Celsius.
assert track4_suggestion["address"] == CELSIUS_URI
assert track4_suggestion["recorded"] == 2
assert track4_suggestion["support"] == Fraction(2, 2) == 1

# The suggestion attaches to the route, not to the Context state.
TRACK4_PREFILL = {
    ("captain-delta", TRACK4_SUGGESTION_CROSSING): track4_suggestion,
}

track4_facets_after = {
    (obs.name, place): uri
    for obs in decidatron_1.members
    for place, uri in obs.facets.items()
}

# The facet stays DARK, and no member's bound facets were mutated.
assert delta_obs.state(TRACK4_DELTA_SPIKE, "meaning") == "DARK"
assert track4_facets_after == track4_facets_before

# The Governor's verdict at the crossing is unchanged by the prefill:
# DARK routes to ASK, now through the suggest-and-confirm Handler.
TRACK4_SUGGEST_RULE = {
    "when": {
        "Context state": "DARK",
        "prefill-available": True,
    },
    "then": {
        "Governor Verdict": "ASK",
        "Decision Route": TRACK4_ROUTES["suggest"],
    },
}

track4_event = {
    "Context state": delta_obs.state(
        TRACK4_DELTA_SPIKE,
        "meaning",
    ),
    "prefill-available": (
        ("captain-delta", TRACK4_SUGGESTION_CROSSING)
        in TRACK4_PREFILL
    ),
}

assert track4_event == TRACK4_SUGGEST_RULE["when"]

track4_instruction = dict(TRACK4_SUGGEST_RULE["then"])

assert track4_instruction == {
    "Governor Verdict": "ASK",
    "Decision Route": TRACK4_ROUTES["suggest"],
}

print_table({
    "crossing": ["/".join(TRACK4_SUGGESTION_CROSSING)],
    "suggested address": [track4_suggestion["address"]],
    "recorded support": [
        f"{track4_suggestion['support'].numerator} of "
        f"{track4_suggestion['recorded']}"
    ],
    "delta's Context state": [
        delta_obs.state(TRACK4_DELTA_SPIKE, "meaning")
    ],
    "Governor Verdict": [track4_instruction["Governor Verdict"]],
    "Decision Route": [track4_instruction["Decision Route"]],
})

print(
    "Verified: the cohort suggestion carries exact recorded support "
    "(2 of 2, Celsius), delta's facet remains DARK, no member's bound "
    "facets changed, and the Governor still routes ASK - now prefilled."
)
````

### **`ASK` as Routed Negotiation: Minting a Referent**

The suggestion machinery above still chooses among *declared* candidates. The last 4-A capability is the one a network of a hundred boundaries actually needs: when participants at the **user level of intent** — the people, above the system and business levels — hold conflicting definitions and **no declared candidate is adequate**, the `ASK` route can carry a negotiation that ends by *minting a new referent*: a locally scoped, versioned `cgvu:` address whose registered entry holds the agreed definition, its scope, its provenance, and its participants. Registration precedes binding; once registered, the address is on the network's vocabulary for others to audit and reuse.

The witness: two captains share the `buoy-7` temperature feed, reading `29`. Epsilon proposes `rounded up`, zeta proposes `rounded down` — a conflict worth a Verdict flip (`NO-GO` vs `GO` on the Atlantic side) — and the sensor's documented behavior, *rounds to nearest*, matches neither declared candidate. Two scripted transcripts keep the cell deterministic. Session A exhausts without agreement: the Governor routes `HALT`, nothing is minted, the facet stays `DARK`. Session B refines against the sensor manual, agrees, registers `…/rounded-to-nearest--north-fleet--v1` — definition, scope, provenance, participants, version — and only then binds: both participants' facets go `LIT` through the same minted address, and the blocked read becomes one deterministic `ACT`.

The honesty rule: **the URI does not create truth.** It gives a negotiated, locally valid definition an address — auditable, reusable, scoped, and versioned. Nothing here claims negotiation always succeeds (session A is the counterexample) or that the minted meaning is universal (its scope and version are part of the record).

````python
# Exact network ASK-negotiation proof: two participants, one contested
# Context Facet, no adequate declared candidate. The ASK route carries a
# scripted negotiation that can fail (HALT, nothing minted) or mint a
# new, locally scoped referent - registered before any binding.

TRACK4_NEGOTIATION_ROUTES = {
    "negotiate": "cgp:/track-4/handlers/negotiate-referent",
    "read": "cgp:/track-4/handlers/read-with-negotiated-referent",
    "refuse": "cgp:/track-4/handlers/refuse-unresolved-referent",
}

NEGOTIATION_INTENT_MAP = {
    "rules": [
        {
            "when": {
                "Context state": "DARK",
                "negotiation": "conflicting-proposals",
            },
            "then": {
                "Governor Verdict": "ASK",
                "Decision Route": TRACK4_NEGOTIATION_ROUTES["negotiate"],
            },
        },
        {
            "when": {
                "Context state": "DARK",
                "negotiation": "exhausted-without-agreement",
            },
            "then": {
                "Governor Verdict": "HALT",
                "Decision Route": TRACK4_NEGOTIATION_ROUTES["refuse"],
            },
        },
        {
            "when": {
                "Context state": "LIT",
                "negotiation": "agreed-and-registered",
            },
            "then": {
                "Governor Verdict": "ACT",
                "Decision Route": TRACK4_NEGOTIATION_ROUTES["read"],
            },
        },
    ],
}


def negotiation_governor(event):
    """Route on the declared rules; anything unresolved stays at ASK."""
    for rule in NEGOTIATION_INTENT_MAP["rules"]:
        if all(
            event.get(key) == value
            for key, value in rule["when"].items()
        ):
            return (
                rule["then"]["Governor Verdict"],
                rule["then"]["Decision Route"],
            )

    return "ASK", TRACK4_NEGOTIATION_ROUTES["negotiate"]


# ACT is structurally prohibited while the facet is DARK: no declared
# rule - and not the default - can route a DARK facet to ACT.
for rule in NEGOTIATION_INTENT_MAP["rules"]:
    if rule["when"]["Context state"] == "DARK":
        assert rule["then"]["Governor Verdict"] != "ACT"


# The shared feed and its two user-level participants.
TRACK4_BUOY7_READING = 29

epsilon_obs = Observatron("captain-epsilon", LOOP_INTENT_MAP)
zeta_obs = Observatron("captain-zeta", LOOP_INTENT_MAP)

for obs in (epsilon_obs, zeta_obs):
    obs.observe("buoy-7", ["temperature"])

TRACK4_BUOY7_SPIKE = 1

assert epsilon_obs.spikes[0]["artifact"] == "buoy-7"
assert zeta_obs.spikes[0]["data"] == "temperature"

# Meaning and world are declared and bound on-vocabulary up front, so
# the contested structure facet isolates the negotiation.
MEANING_URIS = candidate_uris(LOOP_CHOICES, "temperature", "meaning")
WORLD_URIS = candidate_uris(LOOP_CHOICES, "temperature", "world")
STRUCTURE_URIS = candidate_uris(LOOP_CHOICES, "temperature", "structure")

for obs in (epsilon_obs, zeta_obs):
    resolve_facet(obs, TRACK4_BUOY7_SPIKE, "meaning", MEANING_URIS[1])
    resolve_facet(obs, TRACK4_BUOY7_SPIKE, "world", WORLD_URIS[1])
    assert obs.state(TRACK4_BUOY7_SPIKE, "structure") == "DARK"

# The two declared structure candidates are worth a Verdict flip on the
# Atlantic side - this conflict is operational, not cosmetic.
assert decide(TRACK4_BUOY7_READING, False, True, False) == "NO-GO"
assert decide(TRACK4_BUOY7_READING, False, False, False) == "GO"

# ...and the sensor's documented behavior matches neither candidate.
TRACK4_DOCUMENTED_BEHAVIOR = "rounds to nearest whole degree"

assert all(
    TRACK4_DOCUMENTED_BEHAVIOR != choice
    for choice in FACET_CHOICES["temperature.structure"]
)


def negotiate_referent(session):
    """Read one scripted transcript: proposals, conflict, agreement."""
    proposals = {
        key: value
        for key, value in session[0].items()
        if key != "round"
    }

    outcome = {
        "proposals": proposals,
        "conflict": len(set(proposals.values())) > 1,
        "agreed": False,
        "definition": None,
        "rounds": len(session),
    }

    for entry in session[1:]:
        if "agreed-definition" in entry:
            outcome["agreed"] = True
            outcome["definition"] = entry["agreed-definition"]

    return outcome


# SESSION A - the failure path: two rounds, no movement, no agreement.
TRACK4_NEG_SESSION_A = (
    {"round": 1, "captain-epsilon": "rounded up",
     "captain-zeta": "rounded down"},
    {"round": 2, "captain-epsilon": "rounded up",
     "captain-zeta": "rounded down"},
)

track4_session_a = negotiate_referent(TRACK4_NEG_SESSION_A)

assert track4_session_a["conflict"]
assert not track4_session_a["agreed"]

# While unresolved, the Governor holds at ASK...
track4_verdict_open, track4_route_open = negotiation_governor({
    "Context state": epsilon_obs.state(TRACK4_BUOY7_SPIKE, "structure"),
    "negotiation": "conflicting-proposals",
})

assert (track4_verdict_open, track4_route_open) == (
    "ASK",
    TRACK4_NEGOTIATION_ROUTES["negotiate"],
)

# ...and exhaustion without agreement routes HALT: nothing is minted.
track4_verdict_a, track4_route_a = negotiation_governor({
    "Context state": epsilon_obs.state(TRACK4_BUOY7_SPIKE, "structure"),
    "negotiation": "exhausted-without-agreement",
})

assert (track4_verdict_a, track4_route_a) == (
    "HALT",
    TRACK4_NEGOTIATION_ROUTES["refuse"],
)
assert epsilon_obs.state(TRACK4_BUOY7_SPIKE, "structure") == "DARK"
assert zeta_obs.state(TRACK4_BUOY7_SPIKE, "structure") == "DARK"


# SESSION B - the refinement path: cite the documented behavior, agree.
TRACK4_NEG_SESSION_B = (
    {"round": 1, "captain-epsilon": "rounded up",
     "captain-zeta": "rounded down"},
    {"round": 2,
     "refinement-request": "cite the sensor's documented rounding"},
    {"round": 3,
     "captain-epsilon": "manual s4.2: rounds to nearest whole degree",
     "captain-zeta": "manual s4.2: rounds to nearest whole degree"},
    {"round": 4,
     "agreed-definition": (
         "rounded to nearest: the true value lies within half a degree "
         "either side of the shown value; coldest plausible = shown - "
         "0.5 F - the conservative half-step band"
     )},
)

track4_session_b = negotiate_referent(TRACK4_NEG_SESSION_B)

assert track4_session_b["conflict"]
assert track4_session_b["agreed"]
assert "rounded to nearest" in track4_session_b["definition"]


# The minted referent: absent from every initial space, by assertion.
TRACK4_MINTED_URI = (
    "cgvu:/temperature/structure/rounded-to-nearest--north-fleet--v1"
)

assert TRACK4_MINTED_URI not in STRUCTURE_URIS
assert TRACK4_MINTED_URI not in LOOP_URIS
assert TRACK4_MINTED_URI not in decidatron_1.registry

# Binding before registration is refused by the off-vocabulary check.
track4_premature_bind_refused = False

try:
    resolve_facet(
        epsilon_obs,
        TRACK4_BUOY7_SPIKE,
        "structure",
        TRACK4_MINTED_URI,
    )
except AssertionError:
    track4_premature_bind_refused = True

assert track4_premature_bind_refused
assert epsilon_obs.state(TRACK4_BUOY7_SPIKE, "structure") == "DARK"


# Registration: definition, scope, provenance, participants, version -
# then the scoped candidate space and the network registry.
TRACK4_NEGOTIATED_REFERENTS = {
    TRACK4_MINTED_URI: {
        "definition": track4_session_b["definition"],
        "scope": "north-fleet / buoy-7",
        "provenance": "scripted negotiation session B, rounds 1-4",
        "participants": ("captain-epsilon", "captain-zeta"),
        "version": 1,
        "engine-mapping": {"is_rounded_up": True},
        "note": (
            "conservative half-step band - the same band the engine "
            "applies to 'rounded up'"
        ),
    },
}

TRACK4_SCOPED_CANDIDATE_URIS = {
    ("temperature", "structure"): STRUCTURE_URIS + (TRACK4_MINTED_URI,),
}

decidatron_1.registry = tuple(decidatron_1.registry) + (
    TRACK4_MINTED_URI,
)

assert TRACK4_MINTED_URI in decidatron_1.registry
assert TRACK4_MINTED_URI in TRACK4_SCOPED_CANDIDATE_URIS[
    ("temperature", "structure")
]

# The globally declared spaces are untouched: the extension is scoped.
assert LOOP_CHOICES["temperature"]["structure"] == (
    "rounded up",
    "rounded down",
)
assert TRACK4_MINTED_URI not in LOOP_URIS

track4_network_ok, track4_strays = decidatron_1.on_vocabulary()

assert track4_network_ok, track4_strays


def scoped_resolve_facet(obs, spike, facet, uri):
    """resolve_facet against the scope's extended candidate space."""
    data = next(
        record["data"]
        for record in obs.spikes
        if record["spike"] == spike
    )

    allowed = TRACK4_SCOPED_CANDIDATE_URIS.get(
        (data, facet),
        candidate_uris(LOOP_CHOICES, data, facet),
    )

    assert uri in allowed, f"off-vocabulary address: {uri}"

    obs.facets[(spike, facet)] = uri
    return obs.state(spike, facet)


# Binding only after agreement and registration - both participants,
# the same minted address: one referent, reused.
for obs in (epsilon_obs, zeta_obs):
    assert obs.state(TRACK4_BUOY7_SPIKE, "structure") == "DARK"
    scoped_resolve_facet(
        obs,
        TRACK4_BUOY7_SPIKE,
        "structure",
        TRACK4_MINTED_URI,
    )
    assert obs.state(TRACK4_BUOY7_SPIKE, "structure") == "LIT"
    assert obs.uri(TRACK4_BUOY7_SPIKE, "structure") == TRACK4_MINTED_URI

# The previously blocked read is now one deterministic ACT.
track4_verdict_b, track4_route_b = negotiation_governor({
    "Context state": epsilon_obs.state(TRACK4_BUOY7_SPIKE, "structure"),
    "negotiation": "agreed-and-registered",
})

assert (track4_verdict_b, track4_route_b) == (
    "ACT",
    TRACK4_NEGOTIATION_ROUTES["read"],
)

track4_mapping = TRACK4_NEGOTIATED_REFERENTS[TRACK4_MINTED_URI][
    "engine-mapping"
]["is_rounded_up"]

track4_epsilon_read = decide(
    TRACK4_BUOY7_READING, False, track4_mapping, False
)
track4_zeta_read = decide(
    TRACK4_BUOY7_READING, False, track4_mapping, False
)

assert track4_epsilon_read == track4_zeta_read == "NO-GO"

print_table({
    "session": ["A", "B"],
    "rounds": [
        track4_session_a["rounds"],
        track4_session_b["rounds"],
    ],
    "proposals conflict": ["yes", "yes"],
    "agreed": ["no", "yes"],
    "minted": ["-", TRACK4_MINTED_URI],
    "Governor Verdict": [track4_verdict_a, track4_verdict_b],
})

print_table({
    "referent": [TRACK4_MINTED_URI],
    "scope": [
        TRACK4_NEGOTIATED_REFERENTS[TRACK4_MINTED_URI]["scope"]
    ],
    "version": [
        TRACK4_NEGOTIATED_REFERENTS[TRACK4_MINTED_URI]["version"]
    ],
    "participants": [
        ", ".join(
            TRACK4_NEGOTIATED_REFERENTS[TRACK4_MINTED_URI][
                "participants"
            ]
        )
    ],
    "provenance": [
        TRACK4_NEGOTIATED_REFERENTS[TRACK4_MINTED_URI]["provenance"]
    ],
})

print(
    "Verified: conflicting proposals held the Governor at ASK with ACT "
    "prohibited; session A exhausted and routed HALT, minting nothing; "
    "session B agreed, registered the scoped referent before any "
    "binding, both participants went DARK -> LIT through the same "
    "minted address, and the blocked read became one deterministic "
    "NO-GO under ACT."
)
````

## **Track 4-B — Learning Which Questions Are Most Valuable to Ask, and in Which Order**

Section 4-A showed a network can reduce the *cost* of a question. Section 4-B asks the prior question: **which** question, and in what order? Track 2 proved which facets the Verdict depends on; Track 1-C priced misreadings. Section 4-B combines them: bits are shown to be flat by construction, a uniform weighting is declared, the exact expected-regret lattice is computed over all 16 lit-subsets, and the asking order is settled by exhaustion over all 24 permutations — one question worth exactly 49 times the other three, and the right order worth exactly \$1,687,500 per member.

### **Bits Are Flat by Construction**

Every facet key has exactly two declared candidates, so resolving any one facet removes exactly log₂(2) = 1 bit of Hartley uncertainty — no matter which. Four facets, sixteen states, four bits, in any order. **Hartley counting cannot rank the questions.** The ranking must come from what the earlier tracks built: the Decision Shape and the declared payoff ledger.

### The Declared Weighting

Every expectation below weights the finite states uniformly — each writer commitment and each dark reader draw equally. As in Track 1-B: equal likelihood is a controlled assumption of this experiment, not a property of the Context Graph Protocol. Changing the declared weighting changes the numbers; the method — exact expectation over a declared finite set — does not.

````python
# Exact flat-bits proof.
from math import log2

track4_bits_removed = {}

for key in FACET_KEYS:
    assert len(FACET_CHOICES[key]) == 2
    track4_bits_removed[key] = log2(len(FACET_CHOICES[key]))
    assert track4_bits_removed[key] == 1.0

track4_total_states = 2 ** len(FACET_KEYS)

assert track4_total_states == 16
assert log2(track4_total_states) == 4.0

print_table({
    "facet key": list(FACET_KEYS),
    "declared candidates": [
        " / ".join(FACET_CHOICES[key])
        for key in FACET_KEYS
    ],
    "Hartley bits removed": [
        f"{track4_bits_removed[key]:.1f}"
        for key in FACET_KEYS
    ],
})

print(
    "Verified: all four facets remove exactly one bit each - 16 states, "
    "4 bits, in any order. Hartley cannot rank the questions."
)
````

### **The Expected-Regret Lattice**

The next cell computes the exact expected regret per voyage for **every one of the 16 lit-subsets** under the declared weighting — Lean Candidate 4's monotonicity lemma, settled by exhaustion: lighting a facet never increases expected regret; certainty (\$0) requires all four. It also prices each single `ASK` exactly — its EVPI, the expected regret removed by lighting that facet alone — and pins two structural facts: `temperature.meaning` is worth exactly **49 times** any other single facet, and the other three are worth exactly the same amount, an exact three-way tie.

````python
# Candidate 4 - Monotonicity, settled by exhaustion: lighting a facet never
# increases expected regret. Exact expectation over every writer commitment
# and every reader projection, for every one of the 16 possible lit sets.
from itertools import product, combinations
from fractions import Fraction

def exact_expected_regret(lit):
    lit = set(lit)
    dark = [k for k in FACET_KEYS if k not in lit]
    total, n = Fraction(0), 0
    for w_fmt, w_c, w_ru, w_b in product((True, False), (0, 1), (0, 1), (0, 1)):
        for draws in product((0, 1), repeat=len(dark)):
            d = dict(zip(dark, draws))
            r_fmt = w_fmt if "date.structure"        in lit else bool(d["date.structure"])
            r_c   = w_c   if "temperature.meaning"   in lit else d["temperature.meaning"]
            r_ru  = w_ru  if "temperature.structure" in lit else d["temperature.structure"]
            r_b   = w_b   if "temperature.world"     in lit else d["temperature.world"]
            i_true = select_row(db.get(1), QUERY_MONTH, w_fmt)
            i_llm  = select_row(db.get(1), QUERY_MONTH, r_fmt)
            safe    = decide(db.get(1)["temperature"][i_true], w_c, w_ru, w_b)
            captain = decide(db.get(1)["temperature"][i_llm],  r_c, r_ru, r_b)
            total += ledger(captain, safe)[2]
            n += 1
    assert n == 16 * 2 ** len(dark)   # the declared uniform weighting, by construction
    return total / n

E = {frozenset(s): exact_expected_regret(s)
     for k in range(len(FACET_KEYS) + 1)
     for s in combinations(FACET_KEYS, k)}

for S in E:                      # the lemma, over the whole lattice of lit sets
    for T in E:
        if S < T:
            assert E[T] <= E[S], f"lighting {sorted(T - S)} increased expected regret"

assert E[frozenset()] == Fraction(189, 256) * 1_000_000    # $738,281.25 all Dark
assert E[frozenset(FACET_KEYS)] == 0                       # $0 all Lit
assert all(v > 0 for s, v in E.items() if s != frozenset(FACET_KEYS))  # $0 only at all Lit

# Zero expected regret is unique to the all-Lit state.
assert {
    lit_set
    for lit_set, expected_regret in E.items()
    if expected_regret == 0
} == {
    frozenset(FACET_KEYS)
}

ranked = sorted(E.items(), key=lambda kv: (len(kv[0]), -kv[1]))
print_table({
    "lit set": [", ".join(sorted(s)) or "(none - all Dark)" for s, _ in ranked],
    "|lit|":   [len(s) for s, _ in ranked],
    "E[regret] per voyage": [f"${float(v):,.2f}" for _, v in ranked],
})
evpi = {k: E[frozenset()] - E[frozenset([k])] for k in FACET_KEYS}

# Pinned exact values: one dominant question, three exactly tied.
assert evpi["temperature.meaning"] == Fraction(2_296_875, 4)      # $574,218.75
assert (
    evpi["date.structure"]
    == evpi["temperature.structure"]
    == evpi["temperature.world"]
    == Fraction(46_875, 4)                                        # $11,718.75
)
assert evpi["temperature.meaning"] == 49 * evpi["date.structure"]
print("value of lighting one facet (exact EVPI):")
for k, v in sorted(evpi.items(), key=lambda kv: -kv[1]):
    print(f"  {k}: ${float(v):,.2f} per voyage")
print("monotone over all 16 lit sets: lighting never hurts; "
      "certainty ($0) requires all four.")
````

Three exact questions, answered from the lattice: the single best `ASK`; the best *set* of `ASK`s at every budget; and the best *order*, checked against all 4! = 24 permutations under a declared schedule — the fleet flies one voyage-round at every prefix of the order, so an order's total cost per member is the sum of the lattice values along its five prefixes.

````python
# Exact ask-ordering proof: ranking, budgets, and all 24 orders.
from itertools import permutations

# 1 - Single-ASK ranking: one dominant question, three exactly tied.
track4_evpi_ranked = sorted(
    evpi.items(),
    key=lambda kv: (-kv[1], kv[0]),
)

assert track4_evpi_ranked[0][0] == "temperature.meaning"
assert track4_evpi_ranked[0][1] == Fraction(2_296_875, 4)
assert {key for key, _ in track4_evpi_ranked[1:]} == {
    "date.structure",
    "temperature.structure",
    "temperature.world",
}
assert len({value for _, value in track4_evpi_ranked[1:]}) == 1


# 2 - The EVPI-greedy order, and the best subset at every budget.
track4_order, track4_cur = [], frozenset()

while len(track4_cur) < len(FACET_KEYS):
    track4_next = min(
        (key for key in FACET_KEYS if key not in track4_cur),
        key=lambda key: E[track4_cur | {key}],
    )
    track4_order.append(track4_next)
    track4_cur = track4_cur | {track4_next}

assert track4_order[0] == "temperature.meaning"

TRACK4_BUDGET_OPTIMA = {
    0: (Fraction(2_953_125, 4), 1),   # $738,281.25 - all Dark
    1: (Fraction(328_125, 2), 1),     # $164,062.50 - meaning alone
    2: (Fraction(140_625), 3),        # $140,625.00 - a three-way tie
    3: (Fraction(93_750), 3),         # $93,750.00  - a three-way tie
    4: (Fraction(0), 1),              # $0          - all four
}

for budget, (expected_value, expected_ties) in (
    TRACK4_BUDGET_OPTIMA.items()
):
    best_value = min(
        E[subset] for subset in E if len(subset) == budget
    )
    ties = [
        subset
        for subset in E
        if len(subset) == budget and E[subset] == best_value
    ]

    assert best_value == expected_value
    assert len(ties) == expected_ties

    # The greedy prefix is optimal at every budget.
    assert E[frozenset(track4_order[:budget])] == best_value


# 3 - All 24 orders under the declared schedule:
# one voyage-round at every prefix, prefixes 0 through 4.
track4_totals = {
    perm: sum(
        E[frozenset(perm[:i])]
        for i in range(len(FACET_KEYS) + 1)
    )
    for perm in permutations(FACET_KEYS)
}

track4_best_total = min(track4_totals.values())
track4_worst_total = max(track4_totals.values())

assert track4_best_total == Fraction(4_546_875, 4)    # $1,136,718.75 per member
assert track4_worst_total == Fraction(11_296_875, 4)  # $2,824,218.75 per member

track4_optimal_orders = sorted(
    perm
    for perm, total in track4_totals.items()
    if total == track4_best_total
)

track4_worst_orders = [
    perm
    for perm, total in track4_totals.items()
    if total == track4_worst_total
]

# Exactly the six meaning-first orders are optimal; exactly the six
# meaning-last orders are worst; the greedy order is optimal.
assert len(track4_optimal_orders) == 6
assert all(
    perm[0] == "temperature.meaning"
    for perm in track4_optimal_orders
)
assert len(track4_worst_orders) == 6
assert all(
    perm[-1] == "temperature.meaning"
    for perm in track4_worst_orders
)
assert track4_totals[tuple(track4_order)] == track4_best_total

# Order alone is worth exactly $1,687,500 per member across the schedule.
assert track4_worst_total - track4_best_total == Fraction(1_687_500)

print_table({
    "order": [
        " -> ".join(track4_order),
        " -> ".join(reversed(track4_order)),
    ],
    "kind": [
        "EVPI-greedy (optimal)",
        "reverse (worst)",
    ],
    "total per member across 5 rounds": [
        f"${float(track4_best_total):,.2f}",
        f"${float(track4_worst_total):,.2f}",
    ],
})

print(
    "Verified: temperature.meaning first is optimal in all 6 of 24 "
    "optimal orders, meaning-last is worst in all 6 worst orders, the "
    "greedy order is optimal at every budget, and asking in the right "
    "order is worth exactly $1,687,500 per member."
)
````

## **Track 4-C — Amortized Disclosure: The Admiral's Economics**

This section evaluates a homogeneous shared-source network: every member reads the same source and inherits the same recorded resolution. Under that declared assumption, one resolution recorded once can be reused by every member’s route.

Members with different source intents are not covered by this identity. They remain on the confirm-first path demonstrated in Track 4-A, and their aggregate benefit is the sum of their individual benefits.

Section 4-C evaluates two distinct scaling relationships:

1. **Fleet scaling.** At each coverage level, total fleet regret equals the expected regret per voyage from the lattice multiplied by `N_FLEET`. The declared fleet contains `N_FLEET = 256` voyages—one for every complete all-Dark scenario—so every scenario is represented equally and every resulting count is an exact integer.

2. **Membership scaling.** Under the homogeneous shared-source assumption, the value of one recorded and propagated `ASK` equals its per-member EVPI multiplied by `N_MEMBERS`. This assumes that every member inherits the same resolution and that marginal propagation and storage costs are negligible.

Together, these calculations provide the notebook’s minimal finite witness for Lean Candidate 5’s population-monotonicity and amortized-disclosure claim. The section prices the fleet at every coverage level, verifies the exact identity
$$
\text{fleet regret} = \text{expected regret per voyage} \times N_{\text{FLEET}},
$$

and measures the network value of one well-chosen question as

$$
\text{network benefit} = \text{per-member EVPI} \times N_{\text{MEMBERS}}.
$$
The section closes with the Value-of-an-`ASK` visual derived from these exact quantities.


````python
# Exact amortized-disclosure proof: the Admiral's economics.
from itertools import product

N_FLEET = 256  # one voyage per complete all-Dark scenario: exact integers.


def fleet(n_voyages, lit_facets):
    """Exact fleet outcome at one declared lit prefix.

    Returns (misreads, realized cash, optimal cash) over n_voyages,
    assigned evenly across the declared scenario set.
    """
    lit = set(lit_facets)
    dark = [key for key in FACET_KEYS if key not in lit]

    scenario_count = 16 * 2 ** len(dark)
    voyages_per_scenario = n_voyages // scenario_count

    assert voyages_per_scenario * scenario_count == n_voyages

    misreads = realized = optimal = 0

    for w_fmt, w_c, w_ru, w_b in product(
        (True, False), (0, 1), (0, 1), (0, 1)
    ):
        for draws in product((0, 1), repeat=len(dark)):
            d = dict(zip(dark, draws))

            r_fmt = w_fmt if "date.structure" in lit else bool(
                d["date.structure"]
            )
            r_c = w_c if "temperature.meaning" in lit else (
                d["temperature.meaning"]
            )
            r_ru = w_ru if "temperature.structure" in lit else (
                d["temperature.structure"]
            )
            r_b = w_b if "temperature.world" in lit else (
                d["temperature.world"]
            )

            i_true = select_row(db.get(1), QUERY_MONTH, w_fmt)
            i_llm = select_row(db.get(1), QUERY_MONTH, r_fmt)

            safe = decide(
                db.get(1)["temperature"][i_true], w_c, w_ru, w_b
            )
            captain = decide(
                db.get(1)["temperature"][i_llm], r_c, r_ru, r_b
            )

            cash, best, _regret = ledger(captain, safe)

            misreads += voyages_per_scenario * (captain != safe)
            realized += voyages_per_scenario * cash
            optimal += voyages_per_scenario * best

    return misreads, realized, optimal


# The Admiral's table: one fleet per coverage level, EVPI-greedy order.
track4_econ = {
    "lit of 4": [],
    "misreads": [],
    "total regret": [],
    "regret": [],
}
track4_regrets = []
track4_misreads = []

for k in range(len(FACET_KEYS) + 1):
    m, c, o = fleet(N_FLEET, track4_order[:k])

    # Fleet regret equals the lattice, exactly, at every coverage level.
    assert o - c == E[frozenset(track4_order[:k])] * N_FLEET

    track4_regrets.append(o - c)
    track4_misreads.append(m)
    track4_econ["lit of 4"].append(k)
    track4_econ["misreads"].append(f"{m}/{N_FLEET}")
    track4_econ["total regret"].append(f"${o - c:,}")

track4_vmax = max(track4_regrets) or 1
track4_econ["regret"] = [
    hbar(r / track4_vmax, color="#d1242f", width=300)
    for r in track4_regrets
]

print_table(track4_econ)

assert track4_misreads == [126, 28, 24, 16, 0]
assert track4_regrets == [
    189_000_000,
    42_000_000,
    36_000_000,
    24_000_000,
    0,
]

print(
    "same fleet, two gauges: coverage tells the Admiral where the "
    "darkness is; regret tells the Admiral what it costs."
)


# One ASK, propagated: the network inherits the answer via the record.
N_MEMBERS = len(decidatron_1.members)

assert N_MEMBERS == 4

track4_one_ask_gain = (
    E[frozenset()] - E[frozenset(["temperature.meaning"])]
)

assert track4_one_ask_gain == evpi["temperature.meaning"]
assert track4_one_ask_gain > 0

track4_questions_saved = N_MEMBERS - 1

assert track4_questions_saved == 3

print(
    "one ASK, propagated network-wide: "
    f"${float(track4_one_ask_gain):,.2f} saved per member per "
    f"voyage-round x {N_MEMBERS} members = "
    f"${float(track4_one_ask_gain * N_MEMBERS):,.2f} per voyage-round; "
    f"{track4_questions_saved} repeat questions replaced by the record"
)
print(
    "at scale: x 1,000 members = "
    f"${float(track4_one_ask_gain * 1_000):,.2f} per voyage-round, "
    "from a single well-chosen question"
)
print(
    "Verified: fleet regret equals the lattice at every coverage level, "
    "misreads fall 126 -> 28 -> 24 -> 16 -> 0 of 256, and one recorded "
    "ASK is amortized across every member of the network."
)
````

## Candidate 1 — Underdetermination of the artifact

Within the declared model, suppose two admissible writer states produce the same data artifact while assigning different values to a Dark facet. No function of that artifact alone can then recover the correct facet value for both states. Equivalently, the artifact projection is non-injective with respect to the facet.

The result is conditional on the declared answer space and the existence of this collision. It does not claim that every artifact is ambiguous; it establishes that whenever the model admits such a collision, a universally correct artifact-only decoder cannot exist. This is the structural theorem proposed for Lean.

The claim is about the artifact. A reader who nonetheless resolves the facet has
imported a prior — and a prior is not a resolution. It is a second undeclared
facet, held privately, varying by reader, and unrecorded in anything passed
downstream. In cross-jurisdictional settings this is the normal case rather than
the edge case: an analyst in one country, a colleague in another, a client in a
third, subject matter from a fourth, and a source system of unknown provenance.
Each reader's prior is confident and they do not agree. The uniform draw is the
maximum-entropy model of an artifact that resolves nothing.

Priors also compound. A reader who judges the temperature scale by reasoning
about the season has resolved the date facet to get there. One Dark facet is
used to resolve another, and the chain is never surfaced.



## Candidate 2 — Irreducibility of the three Context Facets

The decision modelled in this notebook depends essentially on all three: no
function of only two reproduces it. Each facet has a witness pair — two
interpretations agreeing on the other two, differing in this one, deciding
oppositely.

The lemma is instance-relative and is meant to be. It does not claim every
decision exercises all three facets; it establishes that this one does, which is
enough to rule out the collapse — no two of these facets can stand in for the
third in general, since here they demonstrably cannot. A framework that dropped
one would be unable to express this case.


## Candidate 3 — Non-recoverability along the loop

A reader who misreads a Dark facet and then transmits that projection as intent can erase information about the writer’s original interpretation. Under the downstream observation boundary tested here — the retrieved temperature and final Operational Verdict — no decoder can recover every writer state.

The claim is specific to this declared projection. Additional provenance or operational trace fields may distinguish states that the `(retrieved temperature, final verdict)` pair does not. The finite result therefore establishes non-recoverability at this boundary, not non-recoverability from every possible trace.

The date-format facet selects *which row* is retrieved, so the number read was
itself chosen under someone else's interpretation. Two distinct writer
interpretations can reach the same observable by different routes.

For this finite instance, Candidate 3 requires a collision witness: two distinct writer interpretations that produce the same retrieved value and final decision. The exhaustive check below proves the stronger instance-specific result that all four observable pairs have multiple preimages, with fiber sizes 5, 3, 4, and 4.

At 29 the decision rule reduces to `is_celsius OR (NOT is_baltic AND NOT is_rounded_up)`; at 28 it reduces to `is_celsius`.

## Candidate 5 — Population monotonicity and amortized disclosure

The notebook checks a homogeneous finite instance. It assumes that every registered member inherits the same recorded resolution, faces the same declared scenario distribution, and receives the same per-member reduction in expected regret. Under those assumptions, total benefit is exactly the per-member benefit multiplied by the number of members.

In this instance, lighting `temperature.meaning` once is worth \$574,218.75 per member per voyage-round. The four-member network therefore recovers \$2,296,875.00 per voyage-round, while a 1,000-member homogeneous network recovers \$574,218,750.00.

For heterogeneous members, the general quantity is the sum of the members’ individual benefits; it need not equal one common benefit multiplied by population. The claim that one `ASK` has constant cost also assumes negligible marginal propagation and storage cost.

The Lean target is to prove that if the propagated resolution weakly reduces every member’s regret, then aggregate regret also weakly decreases. Linear `N × benefit` scaling should be stated as a corollary requiring identical per-member benefit.


## Candidate 6 — Gauge invariance under referent relabeling

Within the declared model, let a relabeling be any bijection of the declared `cgvu:` referent vocabulary, applied uniformly to every declared and observed referent — Intent Maps, registries, facet declarations, and resolution records alike. Every gauge observable is then unchanged: the `DARK`/`LIT` state of each Context Facet place, every `MATCH`/`NO MATCH` comparison, all Dark counts, the Essential/Inert certification for the declared Operational Decision, the Operational Verdict, and every regret quantity computed under the declared payoff ledger.

The reason is structural: the declared pipeline consumes referents only through equality tests and registry membership, never through the internal composition of an identifier. Observables therefore factor through the partition that referent equality induces, not through the identifiers themselves. The hypothesis matters — a pipeline that parsed `cgvu:` paths for semantic content would fall outside it — so the claim is relative to opaque referent use, which the executable check is meant to confirm rather than assume.

The recommended executable witness is exhaustion over a generating set: apply every transposition of the declared referent vocabulary and assert that all observables are unchanged. Invariance under transpositions composes to invariance under every permutation, so the finite check covers the full claim for this instance.

The Lean target is the general congruence theorem: gauge observables are well-defined on the quotient of interpretation states by referent renaming — equivalently, the evaluation map is invariant under the action of the symmetric group on the referent vocabulary.

Candidate 6 is complementary to Candidate 3. Candidate 3 establishes that transport around the loop is not path-independent: distinct writer states can collapse to one observable. Candidate 6 establishes what *is* invariant: no observable depends on how a referent happens to be spelled. Together they separate representation, which must not matter, from path, which demonstrably does.

### Discrete gauge-theory formalization direction

A future formalization may model the protocol as a discrete gauge structure over an interaction graph: observed interaction states are vertices; boundary crossings are directed edges; actor-specific interpretations are local frames; declared referent mappings provide transport between local frames; semantics-preserving transformations relate equivalent local representations; protocol measurements that survive those transformations are the gauge-invariant observables of Candidate 6; and composition around a Context Loop tests whether transport is path-independent — the holonomy already exhibited as nontrivial by Track 1-B and Candidate 3.

This is a candidate mathematical formalization. The operational Context Gauge does not depend on completion of that work.

````python
# Addendum exhaustion checks — Candidates 2 and 3.
# This cell settles, by exhaustion over the notebook's own maritime engine
# (`decide`) and the Track 1-C state space, the two claims the addendum marks
# as machine-checked before Lean: the 29-reduction and 28-collapse over all
# 8 belief paths with a verdict-flipping witness pair per facet (Candidate 2),
# and the fiber bound that every downstream observable is consistent with at
# least 3 distinct writer states (Candidate 3).

from itertools import product as _product

# ---- Candidate 3: every observable is consistent with >= 3 writer states ----
_fibers = {k: sorted(v) for k, v in TRACK1C_STATES_BY_VISIBLE_RESULT.items()}
for k, v in _fibers.items():
    print(f"observable {k}: {len(v)} writer states -> {v}")
assert min(len(v) for v in _fibers.values()) >= 3
print("Candidate 3 verified: minimum fiber size is "
      f"{min(len(v) for v in _fibers.values())} over {len(_fibers)} observables.\n")

# ---- Candidate 2: the 29-reduction and 28-collapse over all 8 belief paths ----
for temp, claimed in [
    (29, lambda c, r, b: c or ((not b) and (not r))),
    (28, lambda c, r, b: c),
]:
    for c, r, b in _product((True, False), repeat=3):
        engine = decide(temp, c, r, b)
        formula = "GO" if claimed(c, r, b) else "NO-GO"
        assert engine == formula, (temp, c, r, b, engine, formula)
    print(f"temp={temp}: all 8 belief paths match the claimed reduction.")

# witness pair per facet at 29: flip one facet, hold the other two, verdict flips
for fi, fname in enumerate(("is_celsius", "is_rounded_up", "is_baltic")):
    found = None
    for state in _product((True, False), repeat=3):
        flipped = list(state)
        flipped[fi] = not flipped[fi]
        if decide(29, *state) != decide(29, *flipped):
            found = (state, decide(29, *state), tuple(flipped), decide(29, *flipped))
            break
    assert found, fname
    print(f"witness pair for {fname} at 29: "
          f"{found[0]} -> {found[1]}  vs  {found[2]} -> {found[3]}")

# 28-collapse: structure and world do no work; meaning alone decides
for c in (True, False):
    verdicts = {decide(28, c, r, b) for r, b in _product((True, False), repeat=2)}
    assert len(verdicts) == 1
print("temp=28: for each meaning value, all 4 structure/world combinations agree.")
print("\nCandidates 2 and 3 verified by exhaustion.")
````
